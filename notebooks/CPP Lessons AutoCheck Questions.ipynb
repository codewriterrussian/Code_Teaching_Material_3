````python
# === C++ 自動檢查工具：請先執行這個 cell，不要修改 ===
import base64
import json
import os
import shutil
import subprocess
import tempfile
from IPython.core.magic import register_cell_magic

_TEST_CASES = json.loads(
    base64.b64decode("eyJRMSI6IFt7ImlucHV0IjogIjhcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIwXG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn0sIHsiaW5wdXQiOiAiLTVcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMiI6IFt7ImlucHV0IjogIjE3XG4iLCAiZXhwZWN0ZWQiOiAiUE9TSVRJVkVcbiJ9LCB7ImlucHV0IjogIi05XG4iLCAiZXhwZWN0ZWQiOiAiTkVHQVRJVkVcbiJ9LCB7ImlucHV0IjogIjBcbiIsICJleHBlY3RlZCI6ICJaRVJPXG4ifV0sICJRMyI6IFt7ImlucHV0IjogIjEyXG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiMTBcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICI3XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn1dLCAiUTQiOiBbeyJpbnB1dCI6ICIxNVxuIiwgImV4cGVjdGVkIjogIjFcbiJ9LCB7ImlucHV0IjogIjEwXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiMjFcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRNSI6IFt7ImlucHV0IjogIjdcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIyMFxuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIjI1XG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn1dLCAiUTYiOiBbeyJpbnB1dCI6ICIxNTAwXG4iLCAiZXhwZWN0ZWQiOiAiMTM1MFxuIn0sIHsiaW5wdXQiOiAiOTk5XG4iLCAiZXhwZWN0ZWQiOiAiOTk5XG4ifSwgeyJpbnB1dCI6ICIyMDAwXG4iLCAiZXhwZWN0ZWQiOiAiMTgwMFxuIn1dLCAiUTciOiBbeyJpbnB1dCI6ICIyMCA1XG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiMjAgMFxuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIjIwIDZcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJROCI6IFt7ImlucHV0IjogIjIwIDc1XG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiMTcgOTlcbiIsICJleHBlY3RlZCI6ICIwXG4ifSwgeyJpbnB1dCI6ICIyNSA1MFxuIiwgImV4cGVjdGVkIjogIjBcbiJ9XSwgIlE5IjogW3siaW5wdXQiOiAiMyA0IDVcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIxIDIgM1xuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIjAgNCA0XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn1dLCAiUTEwIjogW3siaW5wdXQiOiAiMyA0IDVcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICI1IDEyIDEzXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiNCA1IDZcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMTEiOiBbeyJpbnB1dCI6ICIxNFxuIiwgImV4cGVjdGVkIjogIkVWRU5cbiJ9LCB7ImlucHV0IjogIjlcbiIsICJleHBlY3RlZCI6ICJPRERcbiJ9LCB7ImlucHV0IjogIjBcbiIsICJleHBlY3RlZCI6ICJFVkVOXG4ifV0sICJRMTIiOiBbeyJpbnB1dCI6ICI4IDNcbiIsICJleHBlY3RlZCI6ICI4XG4ifSwgeyJpbnB1dCI6ICItNSAyXG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiNyA3XG4iLCAiZXhwZWN0ZWQiOiAiN1xuIn1dLCAiUTEzIjogW3siaW5wdXQiOiAiOSAyXG4iLCAiZXhwZWN0ZWQiOiAiR1JFQVRFUlxuIn0sIHsiaW5wdXQiOiAiMyA4XG4iLCAiZXhwZWN0ZWQiOiAiTEVTU1xuIn0sIHsiaW5wdXQiOiAiNSA1XG4iLCAiZXhwZWN0ZWQiOiAiRVFVQUxcbiJ9XSwgIlExNCI6IFt7ImlucHV0IjogIjk1XG4iLCAiZXhwZWN0ZWQiOiAiQVxuIn0sIHsiaW5wdXQiOiAiODRcbiIsICJleHBlY3RlZCI6ICJCXG4ifSwgeyJpbnB1dCI6ICI1OVxuIiwgImV4cGVjdGVkIjogIkZcbiJ9XSwgIlExNSI6IFt7ImlucHV0IjogIjMgOSA1XG4iLCAiZXhwZWN0ZWQiOiAiOVxuIn0sIHsiaW5wdXQiOiAiMTAgLTIgOFxuIiwgImV4cGVjdGVkIjogIjEwXG4ifSwgeyJpbnB1dCI6ICItNCAtNyAtMVxuIiwgImV4cGVjdGVkIjogIi0xXG4ifV0sICJRMTYiOiBbeyJpbnB1dCI6ICIxMiArIDVcbiIsICJleHBlY3RlZCI6ICIxN1xuIn0sIHsiaW5wdXQiOiAiMjAgLyA0XG4iLCAiZXhwZWN0ZWQiOiAiNVxuIn0sIHsiaW5wdXQiOiAiNyAvIDBcbiIsICJleHBlY3RlZCI6ICJJTlZBTElEXG4ifV0sICJRMTciOiBbeyJpbnB1dCI6ICIxXG4iLCAiZXhwZWN0ZWQiOiAiTU9OXG4ifSwgeyJpbnB1dCI6ICI2XG4iLCAiZXhwZWN0ZWQiOiAiU0FUXG4ifSwgeyJpbnB1dCI6ICI5XG4iLCAiZXhwZWN0ZWQiOiAiSU5WQUxJRFxuIn1dLCAiUTE4IjogW3siaW5wdXQiOiAiMlxuIiwgImV4cGVjdGVkIjogIjI4XG4ifSwgeyJpbnB1dCI6ICIxMVxuIiwgImV4cGVjdGVkIjogIjMwXG4ifSwgeyJpbnB1dCI6ICIxM1xuIiwgImV4cGVjdGVkIjogIklOVkFMSURcbiJ9XSwgIlExOSI6IFt7ImlucHV0IjogIkFcbiIsICJleHBlY3RlZCI6ICJWT1dFTFxuIn0sIHsiaW5wdXQiOiAiVVxuIiwgImV4cGVjdGVkIjogIlZPV0VMXG4ifSwgeyJpbnB1dCI6ICJCXG4iLCAiZXhwZWN0ZWQiOiAiT1RIRVJcbiJ9XSwgIlEyMCI6IFt7ImlucHV0IjogIkFcbiIsICJleHBlY3RlZCI6ICIxMDBcbiJ9LCB7ImlucHV0IjogIkNcbiIsICJleHBlY3RlZCI6ICIzMDBcbiJ9LCB7ImlucHV0IjogIlpcbiIsICJleHBlY3RlZCI6ICItMVxuIn1dLCAiUTIxIjogW3siaW5wdXQiOiAiNVxuIiwgImV4cGVjdGVkIjogIjE1XG4ifSwgeyJpbnB1dCI6ICIxMFxuIiwgImV4cGVjdGVkIjogIjU1XG4ifSwgeyJpbnB1dCI6ICIxXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn1dLCAiUTIyIjogW3siaW5wdXQiOiAiNFxuIiwgImV4cGVjdGVkIjogIjRcbjNcbjJcbjFcbiJ9LCB7ImlucHV0IjogIjFcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICI2XG4iLCAiZXhwZWN0ZWQiOiAiNlxuNVxuNFxuM1xuMlxuMVxuIn1dLCAiUTIzIjogW3siaW5wdXQiOiAiNTgzMVxuIiwgImV4cGVjdGVkIjogIjRcbiJ9LCB7ImlucHV0IjogIjBcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIxMDAwMDBcbiIsICJleHBlY3RlZCI6ICI2XG4ifV0sICJRMjQiOiBbeyJpbnB1dCI6ICI1ODMxXG4iLCAiZXhwZWN0ZWQiOiAiMTdcbiJ9LCB7ImlucHV0IjogIjkwMDA5XG4iLCAiZXhwZWN0ZWQiOiAiMThcbiJ9LCB7ImlucHV0IjogIjBcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMjUiOiBbeyJpbnB1dCI6ICI1ODMxXG4iLCAiZXhwZWN0ZWQiOiAiMTM4NVxuIn0sIHsiaW5wdXQiOiAiMTIwMFxuIiwgImV4cGVjdGVkIjogIjIxXG4ifSwgeyJpbnB1dCI6ICI3XG4iLCAiZXhwZWN0ZWQiOiAiN1xuIn1dLCAiUTI2IjogW3siaW5wdXQiOiAiNSA4IDIgLTFcbiIsICJleHBlY3RlZCI6ICIxNVxuIn0sIHsiaW5wdXQiOiAiLTFcbiIsICJleHBlY3RlZCI6ICIwXG4ifSwgeyJpbnB1dCI6ICIxMCAtMyA3IC0xXG4iLCAiZXhwZWN0ZWQiOiAiMTRcbiJ9XSwgIlEyNyI6IFt7ImlucHV0IjogIjEwIDIwIDMwIC0xXG4iLCAiZXhwZWN0ZWQiOiAiMjAuMDBcbiJ9LCB7ImlucHV0IjogIi0xXG4iLCAiZXhwZWN0ZWQiOiAiMC4wMFxuIn0sIHsiaW5wdXQiOiAiNSAtNSAxMCAtMVxuIiwgImV4cGVjdGVkIjogIjMuMzNcbiJ9XSwgIlEyOCI6IFt7ImlucHV0IjogIi0zIDE1IDdcbiIsICJleHBlY3RlZCI6ICI3IDNcbiJ9LCB7ImlucHV0IjogIjVcbiIsICJleHBlY3RlZCI6ICI1IDFcbiJ9LCB7ImlucHV0IjogIjAgMTEgLTggMTBcbiIsICJleHBlY3RlZCI6ICIxMCA0XG4ifV0sICJRMjkiOiBbeyJpbnB1dCI6ICI0XG4iLCAiZXhwZWN0ZWQiOiAiMVxuMlxuM1xuNFxuIn0sIHsiaW5wdXQiOiAiMVxuIiwgImV4cGVjdGVkIjogIjFcbiJ9LCB7ImlucHV0IjogIjZcbiIsICJleHBlY3RlZCI6ICIxXG4yXG4zXG40XG41XG42XG4ifV0sICJRMzAiOiBbeyJpbnB1dCI6ICI3IDMgNSA3XG4iLCAiZXhwZWN0ZWQiOiAiM1xuIn0sIHsiaW5wdXQiOiAiMTAgMTBcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICI0IDEgMiAzIDRcbiIsICJleHBlY3RlZCI6ICI0XG4ifV0sICJRMzEiOiBbeyJpbnB1dCI6ICIxMFxuIiwgImV4cGVjdGVkIjogIjU1XG4ifSwgeyJpbnB1dCI6ICIxXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiMTAwXG4iLCAiZXhwZWN0ZWQiOiAiNTA1MFxuIn1dLCAiUTMyIjogW3siaW5wdXQiOiAiNVxuIiwgImV4cGVjdGVkIjogIjEyMFxuIn0sIHsiaW5wdXQiOiAiMFxuIiwgImV4cGVjdGVkIjogIjFcbiJ9LCB7ImlucHV0IjogIjEwXG4iLCAiZXhwZWN0ZWQiOiAiMzYyODgwMFxuIn1dLCAiUTMzIjogW3siaW5wdXQiOiAiMjAgNFxuIiwgImV4cGVjdGVkIjogIjRcbjhcbjEyXG4xNlxuMjBcbiJ9LCB7ImlucHV0IjogIjE1IDVcbiIsICJleHBlY3RlZCI6ICI1XG4xMFxuMTVcbiJ9LCB7ImlucHV0IjogIjEwIDNcbiIsICJleHBlY3RlZCI6ICIzXG42XG45XG4ifV0sICJRMzQiOiBbeyJpbnB1dCI6ICI1XG4iLCAiZXhwZWN0ZWQiOiAiNVxuNFxuM1xuMlxuMVxuIn0sIHsiaW5wdXQiOiAiMVxuIiwgImV4cGVjdGVkIjogIjFcbiJ9LCB7ImlucHV0IjogIjNcbiIsICJleHBlY3RlZCI6ICIzXG4yXG4xXG4ifV0sICJRMzUiOiBbeyJpbnB1dCI6ICI5XG4iLCAiZXhwZWN0ZWQiOiAiMVxuM1xuNVxuN1xuOVxuIn0sIHsiaW5wdXQiOiAiMTBcbiIsICJleHBlY3RlZCI6ICIxXG4zXG41XG43XG45XG4ifSwgeyJpbnB1dCI6ICIxXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn1dLCAiUTM2IjogW3siaW5wdXQiOiAiMTJcbiIsICJleHBlY3RlZCI6ICI2XG4ifSwgeyJpbnB1dCI6ICIxM1xuIiwgImV4cGVjdGVkIjogIjJcbiJ9LCB7ImlucHV0IjogIjFcbiIsICJleHBlY3RlZCI6ICIxXG4ifV0sICJRMzciOiBbeyJpbnB1dCI6ICIyOVxuIiwgImV4cGVjdGVkIjogIlBSSU1FXG4ifSwgeyJpbnB1dCI6ICIyNVxuIiwgImV4cGVjdGVkIjogIk5PVCBQUklNRVxuIn0sIHsiaW5wdXQiOiAiMVxuIiwgImV4cGVjdGVkIjogIk5PVCBQUklNRVxuIn1dLCAiUTM4IjogW3siaW5wdXQiOiAiMTBcbiIsICJleHBlY3RlZCI6ICIzN1xuIn0sIHsiaW5wdXQiOiAiM1xuIiwgImV4cGVjdGVkIjogIjNcbiJ9LCB7ImlucHV0IjogIjIwXG4iLCAiZXhwZWN0ZWQiOiAiMTQ3XG4ifV0sICJRMzkiOiBbeyJpbnB1dCI6ICJBYkNkZUZcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICJoZWxsb1xuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIlhZWlxuIiwgImV4cGVjdGVkIjogIjNcbiJ9XSwgIlE0MCI6IFt7ImlucHV0IjogIkExQjJDM1xuIiwgImV4cGVjdGVkIjogIjZcbiJ9LCB7ImlucHV0IjogImFiY1xuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIjkweDhcbiIsICJleHBlY3RlZCI6ICIxN1xuIn1dLCAiUTQxIjogW3siaW5wdXQiOiAiMyA1XG4iLCAiZXhwZWN0ZWQiOiAiIyMjIyNcbiMjIyMjXG4jIyMjI1xuIn0sIHsiaW5wdXQiOiAiMSA0XG4iLCAiZXhwZWN0ZWQiOiAiIyMjI1xuIn0sIHsiaW5wdXQiOiAiMiAyXG4iLCAiZXhwZWN0ZWQiOiAiIyNcbiMjXG4ifV0sICJRNDIiOiBbeyJpbnB1dCI6ICI0XG4iLCAiZXhwZWN0ZWQiOiAiI1xuIyNcbiMjI1xuIyMjI1xuIn0sIHsiaW5wdXQiOiAiMVxuIiwgImV4cGVjdGVkIjogIiNcbiJ9LCB7ImlucHV0IjogIjVcbiIsICJleHBlY3RlZCI6ICIjXG4jI1xuIyMjXG4jIyMjXG4jIyMjI1xuIn1dLCAiUTQzIjogW3siaW5wdXQiOiAiNFxuIiwgImV4cGVjdGVkIjogIiMjIyNcbiMjI1xuIyNcbiNcbiJ9LCB7ImlucHV0IjogIjJcbiIsICJleHBlY3RlZCI6ICIjI1xuI1xuIn0sIHsiaW5wdXQiOiAiMVxuIiwgImV4cGVjdGVkIjogIiNcbiJ9XSwgIlE0NCI6IFt7ImlucHV0IjogIjNcbiIsICJleHBlY3RlZCI6ICIxIDIgM1xuMiA0IDZcbjMgNiA5XG4ifSwgeyJpbnB1dCI6ICIxXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiNFxuIiwgImV4cGVjdGVkIjogIjEgMiAzIDRcbjIgNCA2IDhcbjMgNiA5IDEyXG40IDggMTIgMTZcbiJ9XSwgIlE0NSI6IFt7ImlucHV0IjogIjIgM1xuIiwgImV4cGVjdGVkIjogIigxLDEpXG4oMSwyKVxuKDEsMylcbigyLDEpXG4oMiwyKVxuKDIsMylcbiJ9LCB7ImlucHV0IjogIjEgMlxuIiwgImV4cGVjdGVkIjogIigxLDEpXG4oMSwyKVxuIn0sIHsiaW5wdXQiOiAiMyAxXG4iLCAiZXhwZWN0ZWQiOiAiKDEsMSlcbigyLDEpXG4oMywxKVxuIn1dLCAiUTQ2IjogW3siaW5wdXQiOiAiNCA2XG4iLCAiZXhwZWN0ZWQiOiAiIyMjIyMjXG4jLi4uLiNcbiMuLi4uI1xuIyMjIyMjXG4ifSwgeyJpbnB1dCI6ICIyIDVcbiIsICJleHBlY3RlZCI6ICIjIyMjI1xuIyMjIyNcbiJ9LCB7ImlucHV0IjogIjMgM1xuIiwgImV4cGVjdGVkIjogIiMjI1xuIy4jXG4jIyNcbiJ9XSwgIlE0NyI6IFt7ImlucHV0IjogIjVcbiIsICJleHBlY3RlZCI6ICIjLi4uI1xuLiMuIy5cbi4uIy4uXG4uIy4jLlxuIy4uLiNcbiJ9LCB7ImlucHV0IjogIjRcbiIsICJleHBlY3RlZCI6ICIjLi4jXG4uIyMuXG4uIyMuXG4jLi4jXG4ifSwgeyJpbnB1dCI6ICIxXG4iLCAiZXhwZWN0ZWQiOiAiI1xuIn1dLCAiUTQ4IjogW3siaW5wdXQiOiAiNiA3XG4iLCAiZXhwZWN0ZWQiOiAiM1xuIn0sIHsiaW5wdXQiOiAiNSAxMFxuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIjEwIDEyXG4iLCAiZXhwZWN0ZWQiOiAiNFxuIn1dLCAiUTQ5IjogW3siaW5wdXQiOiAiNiA2XG4iLCAiZXhwZWN0ZWQiOiAiNFxuIn0sIHsiaW5wdXQiOiAiMTAgMTJcbiIsICJleHBlY3RlZCI6ICI0XG4ifSwgeyJpbnB1dCI6ICI1IDI1XG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn1dLCAiUTUwIjogW3siaW5wdXQiOiAiNVxuIiwgImV4cGVjdGVkIjogIi4uLi4jXG4uLi4jI1xuLi4jLiNcbi4jLi4jXG4jIyMjI1xuIn0sIHsiaW5wdXQiOiAiM1xuIiwgImV4cGVjdGVkIjogIi4uI1xuLiMjXG4jIyNcbiJ9LCB7ImlucHV0IjogIjJcbiIsICJleHBlY3RlZCI6ICIuI1xuIyNcbiJ9XX0=").decode("utf-8")
)

_CPP_ANSWERS = {}
_RESULTS = {}

def _normalise_output(text):
    lines = [line.rstrip() for line in text.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _compiler():
    for name in ["g++", "clang++", "c++"]:
        path = shutil.which(name)
        if path:
            return path
    return None

def _compile_and_run(source, input_text):
    compiler = _compiler()
    if compiler is None:
        return False, "", "找不到 C++ 編譯器（g++ / clang++ / c++）。"

    with tempfile.TemporaryDirectory() as td:
        src = os.path.join(td, "answer.cpp")
        exe = os.path.join(td, "answer")

        with open(src, "w", encoding="utf-8") as f:
            f.write(source)

        compile_result = subprocess.run(
            [compiler, "-std=c++17", src, "-o", exe],
            capture_output=True,
            text=True,
        )

        if compile_result.returncode != 0:
            return False, "", compile_result.stderr

        try:
            run_result = subprocess.run(
                [exe],
                input=input_text,
                capture_output=True,
                text=True,
                timeout=5,
            )
        except subprocess.TimeoutExpired:
            return False, "", "程式執行超過 5 秒，請檢查是否有無限迴圈。"

        if run_result.returncode != 0:
            err = run_result.stderr or f"程式結束碼：{run_result.returncode}"
            return False, run_result.stdout, err

        return True, run_result.stdout, ""

@register_cell_magic
def cpp_answer(line, cell):
    qid = line.strip().upper()
    _CPP_ANSWERS[qid] = cell
    print(f"已儲存 {qid} 的 C++ 作答。請執行下方檢查 cell。")

def _run_check(qid):
    qid = qid.upper()

    if qid not in _CPP_ANSWERS:
        print(f"❌ {qid}：尚未執行 C++ 作答 cell")
        _RESULTS[qid] = False
        return

    source = _CPP_ANSWERS[qid]
    cases = _TEST_CASES[qid]
    passed = 0

    for case_number, case in enumerate(cases, start=1):
        ok_run, stdout, error = _compile_and_run(source, case["input"])

        if not ok_run:
            print(f"❌ {qid}：編譯或執行失敗")
            print(error)
            _RESULTS[qid] = False
            return

        actual = _normalise_output(stdout)
        expected = _normalise_output(case["expected"])

        if actual == expected:
            passed += 1
        else:
            print(f"❌ {qid} 測試 {case_number} 未通過")
            print("測試輸入：")
            print(case["input"], end="")
            print("你的輸出：")
            print(stdout if stdout else "<沒有輸出>")
            print("預期輸出：")
            print(case["expected"], end="")
            print()

    success = passed == len(cases)
    _RESULTS[qid] = success

    if success:
        print(f"✅ {qid} 全部通過（{passed} / {len(cases)}）")
    else:
        print(f"❌ {qid}：通過 {passed} / {len(cases)} 個測試")

def _final_report():
    passed = sum(_RESULTS.get(f"Q{i}", False) for i in range(1, 51))
    untouched = [f"Q{i}" for i in range(1, 51) if f"Q{i}" not in _RESULTS]

    print(f"總成績：{passed} / 50 題通過")
    if untouched:
        print("尚未執行檢查：" + ", ".join(untouched))

print("C++ 自動檢查工具已載入。")
print("作答後不需要手動輸入測試資料。")
````
