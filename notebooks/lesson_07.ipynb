## Q1

### 測試輸入

```text
10 20 30
```

### C 解法


````python
%%c -n 1
#include <stdio.h>

int main(void) {
    int a, b, c;
    scanf("%d %d %d", &a, &b, &c);

    int temp = a;
    a = b;
    b = c;
    c = temp;

    printf("%d %d %d\n", a, b, c);
    return 0;
}
````

### C++ 解法

````python
%%cpp -n 1
#include <iostream>
using namespace std;

int main() {
    int values[3];

    for (int i = 0; i < 3; i++) {
        cin >> values[i];
    }

    cout << values[1] << " "
         << values[2] << " "
         << values[0] << "\n";

    return 0;
}
````

## Q5

### 測試輸入

```text
80 75 92 88 95
1 1 2 2 4
```

### C 解法


````python
%%c -n 2
#include <stdio.h>

int main(void) {
    double s1, s2, s3, s4, s5;
    double w1, w2, w3, w4, w5;

    scanf(
        "%lf %lf %lf %lf %lf %lf %lf %lf %lf %lf",
        &s1, &s2, &s3, &s4, &s5,
        &w1, &w2, &w3, &w4, &w5
    );

    double total =
        s1 * w1 +
        s2 * w2 +
        s3 * w3 +
        s4 * w4 +
        s5 * w5;

    double weight = w1 + w2 + w3 + w4 + w5;
    double average = total / weight;

    printf("%.2f\n", average);
    return 0;
}
````

### C++ 解法

````python
%%cpp -n 2
#include <iomanip>
#include <iostream>
using namespace std;

int main() {
    double s1, s2, s3, s4, s5;
    double w1, w2, w3, w4, w5;

    cin >> s1 >> s2 >> s3 >> s4 >> s5;
    cin >> w1 >> w2 >> w3 >> w4 >> w5;

    double total =
        s1 * w1 +
        s2 * w2 +
        s3 * w3 +
        s4 * w4 +
        s5 * w5;

    double weight = w1 + w2 + w3 + w4 + w5;
    double average = total / weight;

    cout << fixed << setprecision(2) << average << "\n";
    return 0;
}
````
