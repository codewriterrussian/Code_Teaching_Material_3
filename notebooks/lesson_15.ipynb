# 1. 建立 `vector`

使用 `vector` 前需要：

```cpp
#include <vector>
```

最常見的建立方式：

```cpp
vector<int> a = {10, 20, 30, 40};
```

也可以寫成：

```cpp
vector<int> a{10, 20, 30, 40};
```

兩種寫法在這個例子中效果相同。


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30, 40};
    vector<int> b{10, 20, 30, 40};

    cout << a[0] << endl;
    cout << b[0] << endl;

    return 0;
}
````

## `vector<int>` 是什麼？

```cpp
vector<int> a;
```

代表：

- `vector`：容器
- `<int>`：裡面存放的資料型態
- `a`：變數名稱

因此：

```cpp
vector<int>
```

只能存放整數。

例如也可以建立：

```cpp
vector<double> prices;
vector<string> names;
```


# 2. 指定元素數量

```cpp
vector<int> a(5);
```

代表建立 **5 個 `int` 元素**。

對 `int` 而言，初始值會是 `0`：

```text
0 0 0 0 0
```

概念上有點像 Python：

```python
a = [0] * 5
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a(5);

    for (size_t i = 0; i < a.size(); i++) {
        cout << a[i] << " ";
    }

    return 0;
}
````

如果想建立 5 個元素，而且每一個都設成 `10`：

```cpp
vector<int> a(5, 10);
```

結果：

```text
10 10 10 10 10
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a(5, 10);

    for (size_t i = 0; i < a.size(); i++) {
        cout << a[i] << " ";
    }

    return 0;
}
````

# 3. `size()`：取得元素數量

```cpp
a.size()
```

例如：

```cpp
vector<int> a = {10, 20, 30};
```

則：

```cpp
a.size()
```

得到：


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    cout << a.size() << endl;

    return 0;
}
````

## 固定陣列和 `vector` 的差別

傳統固定陣列：

```cpp
int a[5];
```

本身沒有：

```cpp
a.size()
```

這種成員函式。

對固定陣列，常見的元素數量計算方式是：

```cpp
sizeof(a) / sizeof(a[0])
```

概念是：

```text
整個陣列佔用的記憶體
÷
單一元素佔用的記憶體
=
元素數量
```

例如 5 個 `int`，若每個 `int` 佔 4 bytes：

```text
20 / 4 = 5
```

但對 `vector`，直接使用：

```cpp
a.size()
```

會更方便。

```cpp
a.empty()
```

- 空的 `vector` → `true`
- 有元素 → `false`

例如：

```cpp
vector<int> a;
```

此時：

```cpp
a.empty()
```

為 `true`。


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a;
    vector<int> b = {10, 20, 30};

    cout << boolalpha;

    cout << "a.empty() = " << a.empty() << endl;
    cout << "b.empty() = " << b.empty() << endl;

    return 0;
}
````

# 5. `push_back()`：在最後加入元素

```cpp
a.push_back(40);
```

會把 `40` 放到 `vector` 的最後面。

例如：

```text
10 20 30
```

執行：

```cpp
a.push_back(40);
```

之後：

```text
10 20 30 40
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.push_back(40);
    a.push_back(50);

    for (int x : a) {
        cout << x << " ";
    }

    return 0;
}
````

# 6. `pop_back()`：刪除最後一個元素

```cpp
a.pop_back();
```

會刪除最後一個元素。

例如：

```text
10 20 30
```

執行一次：

```cpp
a.pop_back();
```

變成：

```text
10 20
```

注意：

```cpp
pop_back()
```

**不需要放參數**。


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.pop_back();

    for (int x : a) {
        cout << x << " ";
    }

    return 0;
}
````

# 7. 使用索引走訪 `vector`

和傳統陣列一樣，可以使用：

```cpp
a[i]
```

取得第 `i` 個元素。

常見寫法：

```cpp
for (size_t i = 0; i < a.size(); i++) {
    cout << a[i] << endl;
}
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30, 40, 50};

    for (size_t i = 0; i < a.size(); i++) {
        cout << "a[" << i << "] = " << a[i] << endl;
    }

    return 0;
}
````

## 為什麼使用 `size_t`？

`a.size()` 回傳的型態通常是：

```cpp
size_t
```

因此常寫：

```cpp
for (size_t i = 0; i < a.size(); i++)
```

而不是：

```cpp
for (int i = 0; i < a.size(); i++)
```

後者通常也能執行，但部分編譯器可能出現「signed / unsigned comparison」警告。

初學階段可以先記住：

> 使用索引搭配 `a.size()` 時，`size_t` 是常見且合適的型態。


# 8. Range-based `for`

如果只想依序取得每個元素，而不需要索引，可以使用：

```cpp
for (int x : a) {
    cout << x << endl;
}
```

可以理解成：

> 對 `a` 裡面的每一個元素，依序放到 `x` 中。


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30, 40, 50};

    for (int x : a) {
        cout << x << endl;
    }

    return 0;
}
````

### 索引 `for` 和 Range-based `for`

需要索引時：

```cpp
for (size_t i = 0; i < a.size(); i++) {
    cout << i << " " << a[i] << endl;
}
```

不需要索引時：

```cpp
for (int x : a) {
    cout << x << endl;
}
```

第二種通常比較簡潔。


# 9. `front()`：取得第一個元素

```cpp
a.front()
```

會取得第一個元素。

例如：

```cpp
vector<int> a = {10, 20, 30};
```

則：

```cpp
a.front()
```

得到：

```text
10
```


# 10. `back()`：取得最後一個元素

```cpp
a.back()
```

會取得最後一個元素。

例如：

```cpp
vector<int> a = {10, 20, 30};
```

則：

```cpp
a.back()
```

得到：

```text
30
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30, 40};

    cout << "第一個元素：" << a.front() << endl;
    cout << "最後一個元素：" << a.back() << endl;

    return 0;
}
````

> 注意：如果 `vector` 是空的，不要直接使用 `front()` 或 `back()`。

可以先檢查：

```cpp
if (!a.empty()) {
    cout << a.front();
}
```


# 11. 常用操作整理

| 功能 | 寫法 |
|---|---|
| 建立空 `vector` | `vector<int> a;` |
| 建立並初始化 | `vector<int> a = {10, 20, 30};` |
| 建立 5 個 0 | `vector<int> a(5);` |
| 建立 5 個 10 | `vector<int> a(5, 10);` |
| 元素數量 | `a.size()` |
| 是否為空 | `a.empty()` |
| 最後加入元素 | `a.push_back(x)` |
| 刪除最後元素 | `a.pop_back()` |
| 第一個元素 | `a.front()` |
| 最後一個元素 | `a.back()` |
| 第 `i` 個元素 | `a[i]` |


# 12. 一個完整例子

下面把這些操作串在一起：

```cpp
vector<int> a = {10, 20, 30};

a.push_back(40);
a.push_back(50);
a.pop_back();
```

最後 `a` 會變成：

```text
10 20 30 40
```

因此：

```cpp
a.size()   // 4
a.front()  // 10
a.back()   // 40
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.push_back(40);
    a.push_back(50);
    a.pop_back();

    cout << "size = " << a.size() << endl;
    cout << "front = " << a.front() << endl;
    cout << "back = " << a.back() << endl;

    cout << "內容：";

    for (int x : a) {
        cout << x << " ";
    }

    return 0;
}
````
