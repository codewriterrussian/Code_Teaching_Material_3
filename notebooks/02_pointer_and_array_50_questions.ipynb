````python
# === C / C++ 自動檢查工具：請先執行這個 cell，不要修改 ===
import base64
import json
import os
import shutil
import subprocess
import tempfile
from IPython.core.magic import register_cell_magic

_TEST_CASES = json.loads(
    base64.b64decode("eyJRMSI6IFt7ImlucHV0IjogIjNcbjMgNyA5XG4iLCAiZXhwZWN0ZWQiOiAiM1xuIn0sIHsiaW5wdXQiOiAiMVxuLTVcbiIsICJleHBlY3RlZCI6ICItNVxuIn0sIHsiaW5wdXQiOiAiNFxuMTAgMjAgMzAgNDBcbiIsICJleHBlY3RlZCI6ICIxMFxuIn1dLCAiUTIiOiBbeyJpbnB1dCI6ICIzXG4zIDcgOVxuIiwgImV4cGVjdGVkIjogIjlcbiJ9LCB7ImlucHV0IjogIjFcbi01XG4iLCAiZXhwZWN0ZWQiOiAiLTVcbiJ9LCB7ImlucHV0IjogIjRcbjEwIDIwIDMwIDQwXG4iLCAiZXhwZWN0ZWQiOiAiNDBcbiJ9XSwgIlEzIjogW3siaW5wdXQiOiAiM1xuMSAyIDNcbiIsICJleHBlY3RlZCI6ICI2XG4ifSwgeyJpbnB1dCI6ICIzXG4tMiA1IC0xXG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiMVxuMTBcbiIsICJleHBlY3RlZCI6ICIxMFxuIn1dLCAiUTQiOiBbeyJpbnB1dCI6ICIzXG4xIDIgM1xuIiwgImV4cGVjdGVkIjogIjIuMDBcbiJ9LCB7ImlucHV0IjogIjJcbjIgNVxuIiwgImV4cGVjdGVkIjogIjMuNTBcbiJ9LCB7ImlucHV0IjogIjJcbi0zIDNcbiIsICJleHBlY3RlZCI6ICIwLjAwXG4ifV0sICJRNSI6IFt7ImlucHV0IjogIjRcbjMgOSAyIDdcbiIsICJleHBlY3RlZCI6ICI5XG4ifSwgeyJpbnB1dCI6ICIzXG4tOCAtMiAtNVxuIiwgImV4cGVjdGVkIjogIi0yXG4ifSwgeyJpbnB1dCI6ICIxXG40XG4iLCAiZXhwZWN0ZWQiOiAiNFxuIn1dLCAiUTYiOiBbeyJpbnB1dCI6ICI0XG4zIDkgMiA3XG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiM1xuLTggLTIgLTVcbiIsICJleHBlY3RlZCI6ICItOFxuIn0sIHsiaW5wdXQiOiAiMVxuNFxuIiwgImV4cGVjdGVkIjogIjRcbiJ9XSwgIlE3IjogW3siaW5wdXQiOiAiNFxuMSAtMiAzIDBcbiIsICJleHBlY3RlZCI6ICIyXG4ifSwgeyJpbnB1dCI6ICIyXG4tMSAtMlxuIiwgImV4cGVjdGVkIjogIjBcbiJ9LCB7ImlucHV0IjogIjNcbjUgNiA3XG4iLCAiZXhwZWN0ZWQiOiAiM1xuIn1dLCAiUTgiOiBbeyJpbnB1dCI6ICIzXG4xIDIgM1xuIiwgImV4cGVjdGVkIjogIjIgNCA2XG4ifSwgeyJpbnB1dCI6ICIzXG4tMiAwIDVcbiIsICJleHBlY3RlZCI6ICItNCAwIDEwXG4ifSwgeyJpbnB1dCI6ICIxXG43XG4iLCAiZXhwZWN0ZWQiOiAiMTRcbiJ9XSwgIlE5IjogW3siaW5wdXQiOiAiNFxuMSAyIDMgNFxuIiwgImV4cGVjdGVkIjogIjQgMiAzIDFcbiJ9LCB7ImlucHV0IjogIjFcbjVcbiIsICJleHBlY3RlZCI6ICI1XG4ifSwgeyJpbnB1dCI6ICIyXG4tMSA4XG4iLCAiZXhwZWN0ZWQiOiAiOCAtMVxuIn1dLCAiUTEwIjogW3siaW5wdXQiOiAiM1xuNCA1IDZcbiIsICJleHBlY3RlZCI6ICI0IDUgNlxuIn0sIHsiaW5wdXQiOiAiMlxuLTIgOVxuIiwgImV4cGVjdGVkIjogIi0yIDlcbiJ9LCB7ImlucHV0IjogIjFcbjBcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMTEiOiBbeyJpbnB1dCI6ICIzXG4xIDIgM1xuIiwgImV4cGVjdGVkIjogIjE0XG4ifSwgeyJpbnB1dCI6ICIyXG4tMiA0XG4iLCAiZXhwZWN0ZWQiOiAiMjBcbiJ9LCB7ImlucHV0IjogIjJcbjAgNVxuIiwgImV4cGVjdGVkIjogIjI1XG4ifV0sICJRMTIiOiBbeyJpbnB1dCI6ICIzXG4yIDQgNlxuIiwgImV4cGVjdGVkIjogIjIgNCA2XG4ifSwgeyJpbnB1dCI6ICIyXG4tMSAwXG4iLCAiZXhwZWN0ZWQiOiAiLTEgMFxuIn0sIHsiaW5wdXQiOiAiMVxuOVxuIiwgImV4cGVjdGVkIjogIjlcbiJ9XSwgIlExMyI6IFt7ImlucHV0IjogIjNcbjEgMyA1XG4iLCAiZXhwZWN0ZWQiOiAiMSAzIDVcbiJ9LCB7ImlucHV0IjogIjJcbi03IDJcbiIsICJleHBlY3RlZCI6ICItNyAyXG4ifSwgeyJpbnB1dCI6ICIxXG44XG4iLCAiZXhwZWN0ZWQiOiAiOFxuIn1dLCAiUTE0IjogW3siaW5wdXQiOiAiM1xuNCA2IDhcbiIsICJleHBlY3RlZCI6ICIxOFxuIn0sIHsiaW5wdXQiOiAiMlxuLTQgMVxuIiwgImV4cGVjdGVkIjogIi0zXG4ifSwgeyJpbnB1dCI6ICIyXG4wIDBcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMTUiOiBbeyJpbnB1dCI6ICI1XG4xIDQgNyA0IDlcbjRcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIzXG4yIDMgNVxuOFxuIiwgImV4cGVjdGVkIjogIi0xXG4ifSwgeyJpbnB1dCI6ICIxXG42XG42XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn1dLCAiUTE2IjogW3siaW5wdXQiOiAiNlxuMSAyIDEgMSAzIDJcbjFcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIzXG41IDUgNVxuNVxuIiwgImV4cGVjdGVkIjogIjNcbiJ9LCB7ImlucHV0IjogIjRcbjEgMiAzIDRcbjBcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMTciOiBbeyJpbnB1dCI6ICI1XG4xIDIgMSAzIDFcbjEgOVxuIiwgImV4cGVjdGVkIjogIjkgMiA5IDMgOVxuIn0sIHsiaW5wdXQiOiAiM1xuNCA1IDZcbjcgMFxuIiwgImV4cGVjdGVkIjogIjQgNSA2XG4ifSwgeyJpbnB1dCI6ICIyXG44IDhcbjggLTFcbiIsICJleHBlY3RlZCI6ICItMSAtMVxuIn1dLCAiUTE4IjogW3siaW5wdXQiOiAiNFxuMSAyIDMgNFxuIiwgImV4cGVjdGVkIjogIjIgMyA0IDFcbiJ9LCB7ImlucHV0IjogIjFcbjVcbiIsICJleHBlY3RlZCI6ICI1XG4ifSwgeyJpbnB1dCI6ICIyXG4tMSA5XG4iLCAiZXhwZWN0ZWQiOiAiOSAtMVxuIn1dLCAiUTE5IjogW3siaW5wdXQiOiAiNFxuMSAyIDIgNVxuIiwgImV4cGVjdGVkIjogIjFcbiJ9LCB7ImlucHV0IjogIjNcbjMgMSA0XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn0sIHsiaW5wdXQiOiAiMVxuN1xuIiwgImV4cGVjdGVkIjogIjFcbiJ9XSwgIlEyMCI6IFt7ImlucHV0IjogIjRcbjEgOSAzIDlcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIzXG4tNSAtMiAtOFxuIiwgImV4cGVjdGVkIjogIjFcbiJ9LCB7ImlucHV0IjogIjFcbjRcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMjEiOiBbeyJpbnB1dCI6ICJoZWxsb1xuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogImFcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICJwb2ludGVyXG4iLCAiZXhwZWN0ZWQiOiAiN1xuIn1dLCAiUTIyIjogW3siaW5wdXQiOiAiSGVsbG9cbiIsICJleHBlY3RlZCI6ICIyXG4ifSwgeyJpbnB1dCI6ICJBRUlPVVxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogInJoeXRobVxuIiwgImV4cGVjdGVkIjogIjBcbiJ9XSwgIlEyMyI6IFt7ImlucHV0IjogIkhlbGxvXG4iLCAiZXhwZWN0ZWQiOiAiSEVMTE9cbiJ9LCB7ImlucHV0IjogImFiY1hZWlxuIiwgImV4cGVjdGVkIjogIkFCQ1hZWlxuIn0sIHsiaW5wdXQiOiAiQ1xuIiwgImV4cGVjdGVkIjogIkNcbiJ9XSwgIlEyNCI6IFt7ImlucHV0IjogImhlbGxvXG4iLCAiZXhwZWN0ZWQiOiAib2xsZWhcbiJ9LCB7ImlucHV0IjogImFiY1xuIiwgImV4cGVjdGVkIjogImNiYVxuIn0sIHsiaW5wdXQiOiAieFxuIiwgImV4cGVjdGVkIjogInhcbiJ9XSwgIlEyNSI6IFt7ImlucHV0IjogImxldmVsXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiaGVsbG9cbiIsICJleHBlY3RlZCI6ICIwXG4ifSwgeyJpbnB1dCI6ICJhXG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn1dLCAiUTI2IjogW3siaW5wdXQiOiAiYmFuYW5hIGFcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICJoZWxsbyBsXG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiYWJjIHpcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMjciOiBbeyJpbnB1dCI6ICJiYW5hbmEgblxuIiwgImV4cGVjdGVkIjogIjJcbiJ9LCB7ImlucHV0IjogImFwcGxlIGFcbiIsICJleHBlY3RlZCI6ICIwXG4ifSwgeyJpbnB1dCI6ICJjYXQgelxuIiwgImV4cGVjdGVkIjogIi0xXG4ifV0sICJRMjgiOiBbeyJpbnB1dCI6ICJiYW5hbmEgYSBvXG4iLCAiZXhwZWN0ZWQiOiAiYm9ub25vXG4ifSwgeyJpbnB1dCI6ICJoZWxsbyBsIHhcbiIsICJleHBlY3RlZCI6ICJoZXh4b1xuIn0sIHsiaW5wdXQiOiAiYWJjIHogcVxuIiwgImV4cGVjdGVkIjogImFiY1xuIn1dLCAiUTI5IjogW3siaW5wdXQiOiAiYTFiMjNcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIxMjM0NVxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogImFiY1xuIiwgImV4cGVjdGVkIjogIjBcbiJ9XSwgIlEzMCI6IFt7ImlucHV0IjogImhlbGxvXG4iLCAiZXhwZWN0ZWQiOiAiaCBvXG4ifSwgeyJpbnB1dCI6ICJhYlxuIiwgImV4cGVjdGVkIjogImEgYlxuIn0sIHsiaW5wdXQiOiAiWlxuIiwgImV4cGVjdGVkIjogIlogWlxuIn1dLCAiUTMxIjogW3siaW5wdXQiOiAiMSAyIDNcbiIsICJleHBlY3RlZCI6ICIzIDIgMVxuIn0sIHsiaW5wdXQiOiAiLTEgNSA5XG4iLCAiZXhwZWN0ZWQiOiAiOSA1IC0xXG4ifSwgeyJpbnB1dCI6ICI3IDcgOFxuIiwgImV4cGVjdGVkIjogIjggNyA3XG4ifV0sICJRMzIiOiBbeyJpbnB1dCI6ICIxIDIgMyA0XG4iLCAiZXhwZWN0ZWQiOiAiMTBcbiJ9LCB7ImlucHV0IjogIi0xIDUgMCAyXG4iLCAiZXhwZWN0ZWQiOiAiNlxuIn0sIHsiaW5wdXQiOiAiMTAgMTAgMTAgMTBcbiIsICJleHBlY3RlZCI6ICI0MFxuIn1dLCAiUTMzIjogW3siaW5wdXQiOiAiMSAyIDNcbiIsICJleHBlY3RlZCI6ICIyIDMgNFxuIn0sIHsiaW5wdXQiOiAiLTIgMCA4XG4iLCAiZXhwZWN0ZWQiOiAiLTEgMSA5XG4ifSwgeyJpbnB1dCI6ICI1IDUgNVxuIiwgImV4cGVjdGVkIjogIjYgNiA2XG4ifV0sICJRMzQiOiBbeyJpbnB1dCI6ICI5IDIgNVxuIiwgImV4cGVjdGVkIjogIjIgNSA5XG4ifSwgeyJpbnB1dCI6ICItMSAtOCAzXG4iLCAiZXhwZWN0ZWQiOiAiLTggLTEgM1xuIn0sIHsiaW5wdXQiOiAiNCA0IDFcbiIsICJleHBlY3RlZCI6ICIxIDQgNFxuIn1dLCAiUTM1IjogW3siaW5wdXQiOiAiMSA5IDMgN1xuIiwgImV4cGVjdGVkIjogIjlcbiJ9LCB7ImlucHV0IjogIi01IC0yIC04IC0zXG4iLCAiZXhwZWN0ZWQiOiAiLTJcbiJ9LCB7ImlucHV0IjogIjQgNCA0IDRcbiIsICJleHBlY3RlZCI6ICI0XG4ifV0sICJRMzYiOiBbeyJpbnB1dCI6ICIzIDEgNVxuIiwgImV4cGVjdGVkIjogIjMgMTEgNVxuIn0sIHsiaW5wdXQiOiAiMiAyIDlcbiIsICJleHBlY3RlZCI6ICIxMiAyIDlcbiJ9LCB7ImlucHV0IjogIi04IDAgLTNcbiIsICJleHBlY3RlZCI6ICIyIDAgLTNcbiJ9XSwgIlEzNyI6IFt7ImlucHV0IjogImFwcGxlIGJhbmFuYSBjYXRcbiIsICJleHBlY3RlZCI6ICJhIGIgY1xuIn0sIHsiaW5wdXQiOiAiUmVkIEdyZWVuIEJsdWVcbiIsICJleHBlY3RlZCI6ICJSIEcgQlxuIn0sIHsiaW5wdXQiOiAieCB5eiBxd2VcbiIsICJleHBlY3RlZCI6ICJ4IHkgcVxuIn1dLCAiUTM4IjogW3siaW5wdXQiOiAiYXBwbGUgYmFuYW5hIGNhdFxuIiwgImV4cGVjdGVkIjogIjUgNiAzXG4ifSwgeyJpbnB1dCI6ICJhIGJiIGNjY1xuIiwgImV4cGVjdGVkIjogIjEgMiAzXG4ifSwgeyJpbnB1dCI6ICJwb2ludGVyIGMgY3BwXG4iLCAiZXhwZWN0ZWQiOiAiNyAxIDNcbiJ9XSwgIlEzOSI6IFt7ImlucHV0IjogImNhdCBlbGVwaGFudCBkb2dcbiIsICJleHBlY3RlZCI6ICJlbGVwaGFudFxuIn0sIHsiaW5wdXQiOiAiYWFhIGJiIGNcbiIsICJleHBlY3RlZCI6ICJhYWFcbiJ9LCB7ImlucHV0IjogInggeXogYWJjXG4iLCAiZXhwZWN0ZWQiOiAiYWJjXG4ifV0sICJRNDAiOiBbeyJpbnB1dCI6ICJoZWxsbyB3b3JsZFxuIiwgImV4cGVjdGVkIjogIndvcmxkIGhlbGxvXG4ifSwgeyJpbnB1dCI6ICJhIGJiXG4iLCAiZXhwZWN0ZWQiOiAiYmIgYVxuIn0sIHsiaW5wdXQiOiAicmVkIGJsdWVcbiIsICJleHBlY3RlZCI6ICJibHVlIHJlZFxuIn1dLCAiUTQxIjogW3siaW5wdXQiOiAiNFxuMSAyIDMgNFxuIiwgImV4cGVjdGVkIjogIjEgMyA2IDEwXG4ifSwgeyJpbnB1dCI6ICIzXG4tMSA1IDJcbiIsICJleHBlY3RlZCI6ICItMSA0IDZcbiJ9LCB7ImlucHV0IjogIjFcbjdcbiIsICJleHBlY3RlZCI6ICI3XG4ifV0sICJRNDIiOiBbeyJpbnB1dCI6ICI0XG4xIDIgMyA0XG4iLCAiZXhwZWN0ZWQiOiAiMyA1IDdcbiJ9LCB7ImlucHV0IjogIjJcbjUgLTJcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIzXG4wIDAgMVxuIiwgImV4cGVjdGVkIjogIjAgMVxuIn1dLCAiUTQzIjogW3siaW5wdXQiOiAiM1xuMSAyIDNcbjQgNSA2XG4iLCAiZXhwZWN0ZWQiOiAiMzJcbiJ9LCB7ImlucHV0IjogIjJcbi0xIDJcbjMgNFxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjFcbjdcbjhcbiIsICJleHBlY3RlZCI6ICI1NlxuIn1dLCAiUTQ0IjogW3siaW5wdXQiOiAiNVxuMSAtMiAzIDAgNFxuIiwgImV4cGVjdGVkIjogIjNcbjEgMyA0XG4ifSwgeyJpbnB1dCI6ICIzXG4tMSAwIC01XG4iLCAiZXhwZWN0ZWQiOiAiMFxuRU1QVFlcbiJ9LCB7ImlucHV0IjogIjJcbjYgN1xuIiwgImV4cGVjdGVkIjogIjJcbjYgN1xuIn1dLCAiUTQ1IjogW3siaW5wdXQiOiAiNFxuMyA5IDIgN1xuIiwgImV4cGVjdGVkIjogIjdcbiJ9LCB7ImlucHV0IjogIjNcbi04IC0yIC01XG4iLCAiZXhwZWN0ZWQiOiAiNlxuIn0sIHsiaW5wdXQiOiAiMVxuNFxuIiwgImV4cGVjdGVkIjogIjBcbiJ9XSwgIlE0NiI6IFt7ImlucHV0IjogIjVcbjEgMyAyIDQgNVxuIiwgImV4cGVjdGVkIjogIjNcbiJ9LCB7ImlucHV0IjogIjNcbjMgMiAxXG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn0sIHsiaW5wdXQiOiAiNFxuMSAxIDEgMlxuIiwgImV4cGVjdGVkIjogIjFcbiJ9XSwgIlE0NyI6IFt7ImlucHV0IjogIjRcbjEgMiAzIDRcbjVcbiIsICJleHBlY3RlZCI6ICI2IDIgMyA5XG4ifSwgeyJpbnB1dCI6ICIxXG43XG4zXG4iLCAiZXhwZWN0ZWQiOiAiMTNcbiJ9LCB7ImlucHV0IjogIjJcbi0xIDlcbi0yXG4iLCAiZXhwZWN0ZWQiOiAiLTMgN1xuIn1dLCAiUTQ4IjogW3siaW5wdXQiOiAiNFxuMyA1IDIgN1xuIiwgImV4cGVjdGVkIjogIjEgMyAwIDVcbiJ9LCB7ImlucHV0IjogIjNcbi0yIDAgNFxuIiwgImV4cGVjdGVkIjogIjAgMiA2XG4ifSwgeyJpbnB1dCI6ICIxXG45XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn1dLCAiUTQ5IjogW3siaW5wdXQiOiAiNVxuMSAyIDMgNCA1XG4iLCAiZXhwZWN0ZWQiOiAiNSA0IDMgMiAxXG4ifSwgeyJpbnB1dCI6ICI0XG4tMSAwIDIgOFxuIiwgImV4cGVjdGVkIjogIjggMiAwIC0xXG4ifSwgeyJpbnB1dCI6ICIxXG43XG4iLCAiZXhwZWN0ZWQiOiAiN1xuIn1dLCAiUTUwIjogW3siaW5wdXQiOiAiNVxuMyA5IDIgNyAxXG4iLCAiZXhwZWN0ZWQiOiAiMyAxIDIgNyA5XG4ifSwgeyJpbnB1dCI6ICI0XG41IDUgMSA4XG4iLCAiZXhwZWN0ZWQiOiAiNSA1IDggMVxuIn0sIHsiaW5wdXQiOiAiM1xuNCA0IDRcbiIsICJleHBlY3RlZCI6ICI0IDQgNFxuIn1dfQ==").decode("utf-8")
)

_ANSWERS = {"C": {}, "CPP": {}}
_RESULTS = {"C": {}, "CPP": {}}

def _normalise_output(text):
    lines = [line.rstrip() for line in text.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _compile_and_run(source, input_text, language):
    if language == "C":
        compiler = shutil.which("gcc")
        suffix = ".c"
        flags = ["-std=c11", "-O2"]
    else:
        compiler = shutil.which("g++")
        suffix = ".cpp"
        flags = ["-std=c++17", "-O2"]

    if compiler is None:
        return False, "", f"找不到 {'gcc' if language == 'C' else 'g++'}。請先安裝 C/C++ compiler。"

    with tempfile.TemporaryDirectory() as td:
        src = os.path.join(td, "answer" + suffix)
        exe = os.path.join(td, "answer")
        with open(src, "w", encoding="utf-8") as f:
            f.write(source)

        compile_result = subprocess.run(
            [compiler, src, "-o", exe] + flags,
            capture_output=True,
            text=True,
        )
        if compile_result.returncode != 0:
            return False, "", compile_result.stderr

        try:
            run_result = subprocess.run(
                [exe],
                input=input_text,
                capture_output=True,
                text=True,
                timeout=5,
            )
        except subprocess.TimeoutExpired:
            return False, "", "程式執行超過 5 秒，請檢查是否有無限迴圈。"

        if run_result.returncode != 0:
            err = run_result.stderr or f"程式結束碼：{run_result.returncode}"
            return False, run_result.stdout, err
        return True, run_result.stdout, ""

@register_cell_magic
def c_answer(line, cell):
    qid = line.strip().upper()
    _ANSWERS["C"][qid] = cell
    print(f"已儲存 {qid} 的 C 作答。請執行下方 C 檢查 cell。")

@register_cell_magic
def cpp_answer(line, cell):
    qid = line.strip().upper()
    _ANSWERS["CPP"][qid] = cell
    print(f"已儲存 {qid} 的 C++ 作答。請執行下方 C++ 檢查 cell。")

def _run_check(qid, language):
    qid = qid.upper()
    language = language.upper()
    label = "C" if language == "C" else "C++"

    if qid not in _ANSWERS[language]:
        print(f"❌ {qid}：尚未執行 {label} 作答 cell")
        _RESULTS[language][qid] = False
        return

    source = _ANSWERS[language][qid]
    cases = _TEST_CASES[qid]
    passed = 0

    for case_number, case in enumerate(cases, start=1):
        ok_run, stdout, error = _compile_and_run(source, case["input"], language)
        if not ok_run:
            print(f"❌ {qid}（{label}）：編譯或執行失敗")
            print(error)
            _RESULTS[language][qid] = False
            return

        actual = _normalise_output(stdout)
        expected = _normalise_output(case["expected"])

        if actual == expected:
            passed += 1
        else:
            print(f"❌ {qid}（{label}）測試 {case_number} 未通過")
            print("測試輸入：")
            print(case["input"], end="")
            print("你的輸出：")
            print(stdout if stdout else "<沒有輸出>")
            print("預期輸出：")
            print(case["expected"], end="")
            print()

    success = passed == len(cases)
    _RESULTS[language][qid] = success
    if success:
        print(f"✅ {qid}（{label}）全部通過（{passed} / {len(cases)}）")
    else:
        print(f"❌ {qid}（{label}）：通過 {passed} / {len(cases)} 個測試")

def _language_report(language):
    label = "C" if language == "C" else "C++"
    results = _RESULTS[language]
    passed = sum(results.get(f"Q{i}", False) for i in range(1, 51))
    untouched = [f"Q{i}" for i in range(1, 51) if f"Q{i}" not in results]
    print(f"{label}：{passed} / 50 題通過")
    if untouched:
        print("尚未執行檢查：" + ", ".join(untouched))

def _final_report():
    print("=== C / C++ 總檢查 ===")
    _language_report("C")
    print()
    _language_report("CPP")
    both_passed = sum(
        _RESULTS["C"].get(f"Q{i}", False)
        and _RESULTS["CPP"].get(f"Q{i}", False)
        for i in range(1, 51)
    )
    print()
    print(f"C 與 C++ 都通過：{both_passed} / 50 題")

print("C / C++ 自動檢查工具已載入。")
print("C 使用 C11；C++ 使用 C++17。")
print("每題有 3 組測試資料；作答後不需要手動輸入測試資料。")
````
