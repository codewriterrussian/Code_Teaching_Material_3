### 換行、定位字元與跳脫字元

````python
print("hello\npython")
print("1\t2")
print("\\n")
````

## 2. `input()` 與型態轉換

`input()` 取得的資料型態永遠是字串。需要進行數學計算時，通常要先使用 `int()` 或 `float()` 轉換。

````python
number = int(input("請輸入一個整數："))
print(number + 10)
````

### 一次取得多個輸入

````python
# 例如輸入：apple,banana
first, second = input("請輸入兩個以逗號分隔的內容：").split(",")

print(first)
print(second)
````

`split(",")` 會使用逗號切割字串。左右兩邊的變數數量必須與切割後的資料數量一致。

````python
# 例如輸入：35 4
n, k = map(int, input("請輸入學生總人數與每班人數：").split())

print(n, k)
````

## 3. 字串處理

### `bool()` 的基本規則

常見的假值包括：

- `0`
- `0.0`

非空字串通常為 `True`，即使字串內容是 `"False"`。

## 5. 算術運算子

常用算術運算子：

| 運算子 | 功能 |
|---|---|
| `+` | 加法 |
| `-` | 減法 |
| `*` | 乘法 |
| `/` | 一般除法 |
| `//` | 整數除法 |
| `%` | 餘數 |
| `**` | 次方 |

````python
print(12 ** 5)

print(23 // 5)
print(23 % 5)
````

### 應用：學生分班

輸入學生總人數 `n` 和每班人數 `k`，輸出：

1. 可以分成幾個完整班級
2. 剩下幾位學生

````python
n, k = map(int, input("請輸入學生總人數與每班人數：").split())

class_count = n // k
remaining_students = n % k

print("完整班級數：", class_count)
print("剩餘人數：", remaining_students)
````

## 6. 邏輯運算子

### `and` 真值表

````python
print(True and True)
print(True and False)
print(False and True)
print(False and False)
````

### `and` 與 `or` 的優先順序

Python 會先執行 `and`，再執行 `or`。複雜條件建議加入括號，使執行順序更清楚。

````python
print(True or (False and True))
print((False and False) or True)

result = (
    True
    or (False and True and False)
    or (True and False)
    or (True and True)
)
print(result)

print((1 == "1") or ((1 >= 1) and False) or True)
````

## 7. 比較運算

字串會依照字元順序逐一比較，而不是把內容視為數字。

````python
print("alpha" < "ap")
print("10" < "2")
````

在第二個例子中，字串 `"10"` 小於 `"2"`，因為 Python 先比較第一個字元 `"1"` 和 `"2"`。

## 8. 浮點數精度

````python
result = 0.1 + 0.1 + 0.09 + 0.01

print(result)
print(0.3 == result)
````

電腦使用二進位儲存浮點數，因此部分十進位小數無法被完全精確表示。

需要比較浮點數時，可以比較兩數的差是否足夠小：

```python
abs(0.3 - result) < 1e-9
```

## 9. 輸出格式化

### 控制小數位數

````python
number = 10.3

print("%.2f" % number)
print("%.3f" % number)
print("%.8f" % number)
````

### 控制欄位寬度

````python
print("%8.3f" % 5.0043)
print("%4d" % 5)
````

格式說明：

- `%.2f`：浮點數保留兩位小數。
- `%.3f`：浮點數保留三位小數。
- `%8.3f`：總寬度至少八格，保留三位小數。
- `%4d`：整數至少佔四格。

### 使用 `sep` 格式化姓名

````python
first_name = "John"
second_name = "Mary"

print("Hello", first_name, second_name, sep=", ")
````
