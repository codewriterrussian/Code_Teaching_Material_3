````python
# === C / C++ 自動檢查工具：請先執行這個 cell，不要修改 ===
import base64
import json
import os
import shutil
import subprocess
import tempfile
from IPython.core.magic import register_cell_magic

_TEST_CASES = json.loads(
    base64.b64decode("eyJRMSI6IFt7ImlucHV0IjogIjIgM1xuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIi01IDhcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIwIDBcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMiI6IFt7ImlucHV0IjogIjkgNFxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjIgOFxuIiwgImV4cGVjdGVkIjogIi02XG4ifSwgeyJpbnB1dCI6ICItMyAtN1xuIiwgImV4cGVjdGVkIjogIjRcbiJ9XSwgIlEzIjogW3siaW5wdXQiOiAiMyA3XG4iLCAiZXhwZWN0ZWQiOiAiMjFcbiJ9LCB7ImlucHV0IjogIi00IDVcbiIsICJleHBlY3RlZCI6ICItMjBcbiJ9LCB7ImlucHV0IjogIjAgOVxuIiwgImV4cGVjdGVkIjogIjBcbiJ9XSwgIlE0IjogW3siaW5wdXQiOiAiNCA2XG4iLCAiZXhwZWN0ZWQiOiAiMTBcbiJ9LCB7ImlucHV0IjogIi0yIDVcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIxMCAwXG4iLCAiZXhwZWN0ZWQiOiAiMTBcbiJ9XSwgIlE1IjogW3siaW5wdXQiOiAiNVxuIiwgImV4cGVjdGVkIjogIjI1XG4ifSwgeyJpbnB1dCI6ICItNlxuIiwgImV4cGVjdGVkIjogIjM2XG4ifSwgeyJpbnB1dCI6ICIwXG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn1dLCAiUTYiOiBbeyJpbnB1dCI6ICItOFxuIiwgImV4cGVjdGVkIjogIjhcbiJ9LCB7ImlucHV0IjogIjdcbiIsICJleHBlY3RlZCI6ICI3XG4ifSwgeyJpbnB1dCI6ICIwXG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn1dLCAiUTciOiBbeyJpbnB1dCI6ICIzIDlcbiIsICJleHBlY3RlZCI6ICI5XG4ifSwgeyJpbnB1dCI6ICItMiAtOFxuIiwgImV4cGVjdGVkIjogIi0yXG4ifSwgeyJpbnB1dCI6ICI1IDVcbiIsICJleHBlY3RlZCI6ICI1XG4ifV0sICJROCI6IFt7ImlucHV0IjogIjMgOVxuIiwgImV4cGVjdGVkIjogIjNcbiJ9LCB7ImlucHV0IjogIi0yIC04XG4iLCAiZXhwZWN0ZWQiOiAiLThcbiJ9LCB7ImlucHV0IjogIjUgNVxuIiwgImV4cGVjdGVkIjogIjVcbiJ9XSwgIlE5IjogW3siaW5wdXQiOiAiMCAyIDVcbiIsICJleHBlY3RlZCI6ICI3XG4ifSwgeyJpbnB1dCI6ICIxIDIgNVxuIiwgImV4cGVjdGVkIjogIjEwXG4ifSwgeyJpbnB1dCI6ICI5IC0zIDRcbiIsICJleHBlY3RlZCI6ICItMTJcbiJ9XSwgIlExMCI6IFt7ImlucHV0IjogIjIgM1xuIiwgImV4cGVjdGVkIjogIjUgNlxuIn0sIHsiaW5wdXQiOiAiLTEgNFxuIiwgImV4cGVjdGVkIjogIjMgLTRcbiJ9LCB7ImlucHV0IjogIjAgOFxuIiwgImV4cGVjdGVkIjogIjggMFxuIn1dLCAiUTExIjogW3siaW5wdXQiOiAiMTIgKyA1XG4iLCAiZXhwZWN0ZWQiOiAiMTdcbiJ9LCB7ImlucHV0IjogIjIwIC8gNFxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjcgKiAtM1xuIiwgImV4cGVjdGVkIjogIi0yMVxuIn1dLCAiUTEyIjogW3siaW5wdXQiOiAiMCA1IDNcbiIsICJleHBlY3RlZCI6ICI4XG4ifSwgeyJpbnB1dCI6ICIxIDUgM1xuIiwgImV4cGVjdGVkIjogIjJcbiJ9LCB7ImlucHV0IjogIjIgNSAzXG4iLCAiZXhwZWN0ZWQiOiAiMTVcbiJ9XSwgIlExMyI6IFt7ImlucHV0IjogIjAgNVxuIiwgImV4cGVjdGVkIjogIjI1XG4ifSwgeyJpbnB1dCI6ICIxIC03XG4iLCAiZXhwZWN0ZWQiOiAiN1xuIn0sIHsiaW5wdXQiOiAiMCAtM1xuIiwgImV4cGVjdGVkIjogIjlcbiJ9XSwgIlExNCI6IFt7ImlucHV0IjogIjAgM1xuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjEgM1xuIiwgImV4cGVjdGVkIjogIjEyXG4ifSwgeyJpbnB1dCI6ICIxIC0yXG4iLCAiZXhwZWN0ZWQiOiAiLThcbiJ9XSwgIlExNSI6IFt7ImlucHV0IjogIjAgM1xuIiwgImV4cGVjdGVkIjogIjhcbiJ9LCB7ImlucHV0IjogIjEgM1xuIiwgImV4cGVjdGVkIjogIjdcbiJ9LCB7ImlucHV0IjogIjAgLTFcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMTYiOiBbeyJpbnB1dCI6ICIwIDggMlxuIiwgImV4cGVjdGVkIjogIjEwXG4ifSwgeyJpbnB1dCI6ICIxIDggMlxuIiwgImV4cGVjdGVkIjogIjZcbiJ9LCB7ImlucHV0IjogIjIgOCAyXG4iLCAiZXhwZWN0ZWQiOiAiMTZcbiJ9XSwgIlExNyI6IFt7ImlucHV0IjogIjAgMSAyIDNcbiIsICJleHBlY3RlZCI6ICI2XG4ifSwgeyJpbnB1dCI6ICIxIDEwIDMgMlxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjIgMiAzIDRcbiIsICJleHBlY3RlZCI6ICIyNFxuIn1dLCAiUTE4IjogW3siaW5wdXQiOiAiMCAtOCAzXG4iLCAiZXhwZWN0ZWQiOiAiNVxuIn0sIHsiaW5wdXQiOiAiMSAzIDhcbiIsICJleHBlY3RlZCI6ICI1XG4ifSwgeyJpbnB1dCI6ICIxIDEwIDJcbiIsICJleHBlY3RlZCI6ICI4XG4ifV0sICJRMTkiOiBbeyJpbnB1dCI6ICIwIDEgMiAzXG4iLCAiZXhwZWN0ZWQiOiAiNlxuIn0sIHsiaW5wdXQiOiAiMSAxIDggM1xuIiwgImV4cGVjdGVkIjogIjhcbiJ9LCB7ImlucHV0IjogIjEgLTIgLTUgLTFcbiIsICJleHBlY3RlZCI6ICItMVxuIn1dLCAiUTIwIjogW3siaW5wdXQiOiAiMCA4XG4iLCAiZXhwZWN0ZWQiOiAiMVxuIn0sIHsiaW5wdXQiOiAiMCA3XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn0sIHsiaW5wdXQiOiAiMSAtM1xuIiwgImV4cGVjdGVkIjogIjBcbiJ9XSwgIlEyMSI6IFt7ImlucHV0IjogIjNcbjEgMiAzXG4iLCAiZXhwZWN0ZWQiOiAiMSA0IDlcbiJ9LCB7ImlucHV0IjogIjNcbi0yIDAgNFxuIiwgImV4cGVjdGVkIjogIjQgMCAxNlxuIn0sIHsiaW5wdXQiOiAiMVxuNVxuIiwgImV4cGVjdGVkIjogIjI1XG4ifV0sICJRMjIiOiBbeyJpbnB1dCI6ICI0XG4tMSAyIC0zIDBcbiIsICJleHBlY3RlZCI6ICIxIDIgMyAwXG4ifSwgeyJpbnB1dCI6ICIyXG4tNSAtNlxuIiwgImV4cGVjdGVkIjogIjUgNlxuIn0sIHsiaW5wdXQiOiAiMVxuN1xuIiwgImV4cGVjdGVkIjogIjdcbiJ9XSwgIlEyMyI6IFt7ImlucHV0IjogIjBcbjNcbjEgMiAzXG4iLCAiZXhwZWN0ZWQiOiAiMTRcbiJ9LCB7ImlucHV0IjogIjFcbjNcbi0xIDIgLTNcbiIsICJleHBlY3RlZCI6ICI2XG4ifSwgeyJpbnB1dCI6ICIwXG4yXG4tMiA0XG4iLCAiZXhwZWN0ZWQiOiAiMjBcbiJ9XSwgIlEyNCI6IFt7ImlucHV0IjogIjVcbjEgMiA0IDcgOFxuIiwgImV4cGVjdGVkIjogIjNcbiJ9LCB7ImlucHV0IjogIjNcbjEgMyA1XG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn0sIHsiaW5wdXQiOiAiMlxuMCAtMlxuIiwgImV4cGVjdGVkIjogIjJcbiJ9XSwgIlEyNSI6IFt7ImlucHV0IjogIjVcbjEgLTIgMyAwIDRcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIzXG4tMSAwIC0yXG4iLCAiZXhwZWN0ZWQiOiAiMFxuIn0sIHsiaW5wdXQiOiAiMlxuNiA3XG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn1dLCAiUTI2IjogW3siaW5wdXQiOiAiMCAzIDEgMlxuIiwgImV4cGVjdGVkIjogIjEgMiAzXG4ifSwgeyJpbnB1dCI6ICIxIDMgMSAyXG4iLCAiZXhwZWN0ZWQiOiAiMyAyIDFcbiJ9LCB7ImlucHV0IjogIjAgLTEgLTUgNFxuIiwgImV4cGVjdGVkIjogIi01IC0xIDRcbiJ9XSwgIlEyNyI6IFt7ImlucHV0IjogIjBcbjVcbjEgMiAzIDQgNVxuIiwgImV4cGVjdGVkIjogIjIgNFxuIn0sIHsiaW5wdXQiOiAiMVxuNFxuLTIgMCAzIDVcbiIsICJleHBlY3RlZCI6ICIzIDVcbiJ9LCB7ImlucHV0IjogIjFcbjJcbi0yIDBcbiIsICJleHBlY3RlZCI6ICJFTVBUWVxuIn1dLCAiUTI4IjogW3siaW5wdXQiOiAiMFxuNVxuMSAzIDQgNiA3XG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiMVxuM1xuLTIgMCA5XG4iLCAiZXhwZWN0ZWQiOiAiMlxuIn0sIHsiaW5wdXQiOiAiMFxuM1xuMSAzIDVcbiIsICJleHBlY3RlZCI6ICItMVxuIn1dLCAiUTI5IjogW3siaW5wdXQiOiAiMFxuM1xuMSAyIDNcbiIsICJleHBlY3RlZCI6ICIyIDMgNFxuIn0sIHsiaW5wdXQiOiAiMVxuM1xuMSAyIDNcbiIsICJleHBlY3RlZCI6ICIyIDQgNlxuIn0sIHsiaW5wdXQiOiAiMVxuMlxuLTIgMFxuIiwgImV4cGVjdGVkIjogIi00IDBcbiJ9XSwgIlEzMCI6IFt7ImlucHV0IjogIjAgMS41IDIuMjVcbiIsICJleHBlY3RlZCI6ICIzLjc1XG4ifSwgeyJpbnB1dCI6ICIxIDUuMCAxLjI1XG4iLCAiZXhwZWN0ZWQiOiAiMy43NVxuIn0sIHsiaW5wdXQiOiAiMiAxLjUgMi4wXG4iLCAiZXhwZWN0ZWQiOiAiMy4wMFxuIn1dLCAiUTMxIjogW3siaW5wdXQiOiAiMiA4XG4iLCAiZXhwZWN0ZWQiOiAiMTBcbiJ9LCB7ImlucHV0IjogIi00IDdcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIwIDBcbiIsICJleHBlY3RlZCI6ICIwXG4ifV0sICJRMzIiOiBbeyJpbnB1dCI6ICIwIDMgOVxuIiwgImV4cGVjdGVkIjogIjNcbiJ9LCB7ImlucHV0IjogIjEgMyA5XG4iLCAiZXhwZWN0ZWQiOiAiOVxuIn0sIHsiaW5wdXQiOiAiMSAtMiAtOFxuIiwgImV4cGVjdGVkIjogIi0yXG4ifV0sICJRMzMiOiBbeyJpbnB1dCI6ICIwIDcgMlxuIiwgImV4cGVjdGVkIjogIjlcbiJ9LCB7ImlucHV0IjogIjEgNyAyXG4iLCAiZXhwZWN0ZWQiOiAiNVxuIn0sIHsiaW5wdXQiOiAiMiA3IDJcbiIsICJleHBlY3RlZCI6ICIxNFxuIn1dLCAiUTM0IjogW3siaW5wdXQiOiAiMCA1XG4iLCAiZXhwZWN0ZWQiOiAiMjVcbiJ9LCB7ImlucHV0IjogIjEgNVxuIiwgImV4cGVjdGVkIjogIjZcbiJ9LCB7ImlucHV0IjogIjAgLTNcbiIsICJleHBlY3RlZCI6ICI5XG4ifV0sICJRMzUiOiBbeyJpbnB1dCI6ICIwIDggM1xuIiwgImV4cGVjdGVkIjogIjhcbiJ9LCB7ImlucHV0IjogIjEgOCAzXG4iLCAiZXhwZWN0ZWQiOiAiM1xuIn0sIHsiaW5wdXQiOiAiMSAtNSAtMlxuIiwgImV4cGVjdGVkIjogIi01XG4ifV0sICJRMzYiOiBbeyJpbnB1dCI6ICIwIDVcbiIsICJleHBlY3RlZCI6ICI2XG4ifSwgeyJpbnB1dCI6ICIxIDVcbiIsICJleHBlY3RlZCI6ICI0XG4ifSwgeyJpbnB1dCI6ICIxIC0yXG4iLCAiZXhwZWN0ZWQiOiAiLTNcbiJ9XSwgIlEzNyI6IFt7ImlucHV0IjogIjEgOVxuIiwgImV4cGVjdGVkIjogIjkgMVxuIn0sIHsiaW5wdXQiOiAiLTIgNFxuIiwgImV4cGVjdGVkIjogIjQgLTJcbiJ9LCB7ImlucHV0IjogIjUgNVxuIiwgImV4cGVjdGVkIjogIjUgNVxuIn1dLCAiUTM4IjogW3siaW5wdXQiOiAiMCA1XG4iLCAiZXhwZWN0ZWQiOiAiNlxuIn0sIHsiaW5wdXQiOiAiMSA1XG4iLCAiZXhwZWN0ZWQiOiAiNFxuIn0sIHsiaW5wdXQiOiAiMiA1XG4iLCAiZXhwZWN0ZWQiOiAiMTBcbiJ9XSwgIlEzOSI6IFt7ImlucHV0IjogIjAgMiAzXG4iLCAiZXhwZWN0ZWQiOiAiOFxuIn0sIHsiaW5wdXQiOiAiMiAxIDNcbiIsICJleHBlY3RlZCI6ICI1XG4ifSwgeyJpbnB1dCI6ICIxIDEgMFxuIiwgImV4cGVjdGVkIjogIi0yXG4ifV0sICJRNDAiOiBbeyJpbnB1dCI6ICIwIDIgMSAzXG4iLCAiZXhwZWN0ZWQiOiAiN1xuIn0sIHsiaW5wdXQiOiAiMiAyIDAgMVxuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjEgMCAwIDRcbiIsICJleHBlY3RlZCI6ICI1XG4ifV0sICJRNDEiOiBbeyJpbnB1dCI6ICIwIGFcbiIsICJleHBlY3RlZCI6ICJBXG4ifSwgeyJpbnB1dCI6ICIxIFpcbiIsICJleHBlY3RlZCI6ICJ6XG4ifSwgeyJpbnB1dCI6ICIwIFFcbiIsICJleHBlY3RlZCI6ICJRXG4ifV0sICJRNDIiOiBbeyJpbnB1dCI6ICIwIGFcbiIsICJleHBlY3RlZCI6ICIxXG4ifSwgeyJpbnB1dCI6ICIwIHpcbiIsICJleHBlY3RlZCI6ICIwXG4ifSwgeyJpbnB1dCI6ICIxIDdcbiIsICJleHBlY3RlZCI6ICIxXG4ifV0sICJRNDMiOiBbeyJpbnB1dCI6ICIwIEhlbGxvXG4iLCAiZXhwZWN0ZWQiOiAiSEVMTE9cbiJ9LCB7ImlucHV0IjogIjEgQWJDXG4iLCAiZXhwZWN0ZWQiOiAiYWJjXG4ifSwgeyJpbnB1dCI6ICIwIHh5elxuIiwgImV4cGVjdGVkIjogIlhZWlxuIn1dLCAiUTQ0IjogW3siaW5wdXQiOiAiMCAzIDkgNFxuIiwgImV4cGVjdGVkIjogIjlcbiJ9LCB7ImlucHV0IjogIjEgMyA5IDRcbiIsICJleHBlY3RlZCI6ICIzXG4ifSwgeyJpbnB1dCI6ICIxIC0xIC04IDJcbiIsICJleHBlY3RlZCI6ICItOFxuIn1dLCAiUTQ1IjogW3siaW5wdXQiOiAiMFxuNFxuMSAyIDMgNFxuIiwgImV4cGVjdGVkIjogIjEwXG4ifSwgeyJpbnB1dCI6ICIxXG40XG4xIDIgMyA0XG4iLCAiZXhwZWN0ZWQiOiAiMjRcbiJ9LCB7ImlucHV0IjogIjFcbjNcbi0xIDIgM1xuIiwgImV4cGVjdGVkIjogIi02XG4ifV0sICJRNDYiOiBbeyJpbnB1dCI6ICIwIDggMlxuIiwgImV4cGVjdGVkIjogIjEwXG4ifSwgeyJpbnB1dCI6ICIyIDggMlxuIiwgImV4cGVjdGVkIjogIjE2XG4ifSwgeyJpbnB1dCI6ICIzIDkgMlxuIiwgImV4cGVjdGVkIjogIjRcbiJ9XSwgIlE0NyI6IFt7ImlucHV0IjogIjBcbjNcbjEgMiAzXG4iLCAiZXhwZWN0ZWQiOiAiMTEgMTIgMTNcbiJ9LCB7ImlucHV0IjogIjFcbjNcbjEgMiAzXG4iLCAiZXhwZWN0ZWQiOiAiMiA0IDZcbiJ9LCB7ImlucHV0IjogIjFcbjJcbi0yIDBcbiIsICJleHBlY3RlZCI6ICItNCAwXG4ifV0sICJRNDgiOiBbeyJpbnB1dCI6ICJBIDggM1xuIiwgImV4cGVjdGVkIjogIjExXG4ifSwgeyJpbnB1dCI6ICJTIDggM1xuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIk0gOCAzXG4iLCAiZXhwZWN0ZWQiOiAiMjRcbiJ9XSwgIlE0OSI6IFt7ImlucHV0IjogIjAgM1xuIiwgImV4cGVjdGVkIjogIjVcbiJ9LCB7ImlucHV0IjogIjEgM1xuIiwgImV4cGVjdGVkIjogIjEyXG4ifSwgeyJpbnB1dCI6ICIxIC0yXG4iLCAiZXhwZWN0ZWQiOiAiLThcbiJ9XSwgIlE1MCI6IFt7ImlucHV0IjogIjAgMFxuNFxuLTEgMiAtMyA0XG4iLCAiZXhwZWN0ZWQiOiAiMTBcbiJ9LCB7ImlucHV0IjogIjEgMFxuM1xuMSAyIDNcbiIsICJleHBlY3RlZCI6ICIxNFxuIn0sIHsiaW5wdXQiOiAiMCAxXG40XG4tOCAyIC0zIDRcbiIsICJleHBlY3RlZCI6ICI4XG4ifV19").decode("utf-8")
)

_ANSWERS = {"C": {}, "CPP": {}}
_RESULTS = {"C": {}, "CPP": {}}

def _normalise_output(text):
    lines = [line.rstrip() for line in text.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _compile_and_run(source, input_text, language):
    if language == "C":
        compiler = shutil.which("gcc")
        suffix = ".c"
        flags = ["-std=c11", "-O2"]
    else:
        compiler = shutil.which("g++")
        suffix = ".cpp"
        flags = ["-std=c++17", "-O2"]

    if compiler is None:
        return False, "", f"找不到 {'gcc' if language == 'C' else 'g++'}。請先安裝 C/C++ compiler。"

    with tempfile.TemporaryDirectory() as td:
        src = os.path.join(td, "answer" + suffix)
        exe = os.path.join(td, "answer")
        with open(src, "w", encoding="utf-8") as f:
            f.write(source)

        compile_result = subprocess.run(
            [compiler, src, "-o", exe] + flags,
            capture_output=True,
            text=True,
        )
        if compile_result.returncode != 0:
            return False, "", compile_result.stderr

        try:
            run_result = subprocess.run(
                [exe],
                input=input_text,
                capture_output=True,
                text=True,
                timeout=5,
            )
        except subprocess.TimeoutExpired:
            return False, "", "程式執行超過 5 秒，請檢查是否有無限迴圈。"

        if run_result.returncode != 0:
            err = run_result.stderr or f"程式結束碼：{run_result.returncode}"
            return False, run_result.stdout, err
        return True, run_result.stdout, ""

@register_cell_magic
def c_answer(line, cell):
    qid = line.strip().upper()
    _ANSWERS["C"][qid] = cell
    print(f"已儲存 {qid} 的 C 作答。請執行下方 C 檢查 cell。")

@register_cell_magic
def cpp_answer(line, cell):
    qid = line.strip().upper()
    _ANSWERS["CPP"][qid] = cell
    print(f"已儲存 {qid} 的 C++ 作答。請執行下方 C++ 檢查 cell。")

def _run_check(qid, language):
    qid = qid.upper()
    language = language.upper()
    label = "C" if language == "C" else "C++"

    if qid not in _ANSWERS[language]:
        print(f"❌ {qid}：尚未執行 {label} 作答 cell")
        _RESULTS[language][qid] = False
        return

    source = _ANSWERS[language][qid]
    cases = _TEST_CASES[qid]
    passed = 0

    for case_number, case in enumerate(cases, start=1):
        ok_run, stdout, error = _compile_and_run(source, case["input"], language)
        if not ok_run:
            print(f"❌ {qid}（{label}）：編譯或執行失敗")
            print(error)
            _RESULTS[language][qid] = False
            return

        actual = _normalise_output(stdout)
        expected = _normalise_output(case["expected"])

        if actual == expected:
            passed += 1
        else:
            print(f"❌ {qid}（{label}）測試 {case_number} 未通過")
            print("測試輸入：")
            print(case["input"], end="")
            print("你的輸出：")
            print(stdout if stdout else "<沒有輸出>")
            print("預期輸出：")
            print(case["expected"], end="")
            print()

    success = passed == len(cases)
    _RESULTS[language][qid] = success
    if success:
        print(f"✅ {qid}（{label}）全部通過（{passed} / {len(cases)}）")
    else:
        print(f"❌ {qid}（{label}）：通過 {passed} / {len(cases)} 個測試")

def _language_report(language):
    label = "C" if language == "C" else "C++"
    results = _RESULTS[language]
    passed = sum(results.get(f"Q{i}", False) for i in range(1, 51))
    untouched = [f"Q{i}" for i in range(1, 51) if f"Q{i}" not in results]
    print(f"{label}：{passed} / 50 題通過")
    if untouched:
        print("尚未執行檢查：" + ", ".join(untouched))

def _final_report():
    print("=== C / C++ 總檢查 ===")
    _language_report("C")
    print()
    _language_report("CPP")
    both_passed = sum(
        _RESULTS["C"].get(f"Q{i}", False)
        and _RESULTS["CPP"].get(f"Q{i}", False)
        for i in range(1, 51)
    )
    print()
    print(f"C 與 C++ 都通過：{both_passed} / 50 題")

print("C / C++ 自動檢查工具已載入。")
print("C 使用 C11；C++ 使用 C++17。")
print("每題有 3 組測試資料；作答後不需要手動輸入測試資料。")
````
