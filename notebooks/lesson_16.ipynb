```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int x = 10;
    int *p = &x;

    *p = 50;

    cout << x << endl;

    return 0;
}
```

表達式	意義
x	x 的值
&x	x 的位址
p	p 裡儲存的位址
*p	p 所指向位置的值
&p	指標變數 p 自己的位址

|x| ..... |p (->x)|

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int x = 100;
    int *t = &x;

    cout << "x = " << x << endl;
    cout << "&x = " << &x << endl;
    cout << "t = " << t << endl;
    cout << "*t = " << *t << endl;
    cout << "&t = " << &t << endl;

    return 0;
}
```
