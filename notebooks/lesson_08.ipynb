## 2. C++：使用參考參數取得多個結果

函式：

```cpp
void calculate(int a, int b, int& sum, int& product)
```

接收四個參數。

其中：

```cpp
int a
int b
```

是一般的傳值參數。

而：

```cpp
int& sum
int& product
```

是參考參數，可以直接修改呼叫端的變數。


````python
%%cpp

#include <iostream>
using namespace std;

void calculate(int a, int b, int& sum, int& product) {
    sum = a + b;
    product = a * b;
}

int main() {
    int s, p;

    calculate(3, 4, s, p);

    cout << s << endl;
    cout << p << endl;

    return 0;
}
````

### 呼叫流程

呼叫：

```cpp
calculate(3, 4, s, p);
```

可以理解成：

```text
a = 3
b = 4

sum     → 參考 s
product → 參考 p
```

函式內：

```cpp
sum = a + b;
```

因此：

```text
s = 3 + 4
  = 7
```

接著：

```cpp
product = a * b;
```

因此：

```text
p = 3 * 4
  = 12
```

最後：

```cpp
cout << s << endl;
cout << p << endl;
```

會輸出：

```text
7
12
```


### 為什麼使用 `int&`？

如果函式宣告成：

```cpp
void calculate(int a, int b, int sum, int product)
```

`sum` 與 `product` 只會是呼叫端資料的複本。

改成：

```cpp
int& sum
int& product
```

後，函式內對它們的修改會直接影響：

```cpp
s
p
```

因此，一個 `void` 函式也可以透過參考參數帶回多個結果。


## 3. C：複合指定運算子 `+=`

在 C 中：

```c
a += 1;
```

表示把 `a` 原本的值加上 `1`，再存回 `a`。


````python
%%c

#include <stdio.h>

int main(void) {
    int a = 2;

    a += 1;

    printf("%d\n", a);

    return 0;
}
````

### `+=` 的意思

下面兩種寫法具有相同的效果：

```c
a += 1;
```

與：

```c
a = a + 1;
```

在原始程式中：

```text
a = 2
```

執行：

```c
a += 1;
```

之後：

```text
a = 3
```

所以：

```c
printf("%d\n", a);
```

會輸出：

```text
3
```


## 4. 重點整理

### C++ 參考參數

```cpp
int& value
```

表示參數會參考呼叫端的原始變數。

例如：

```cpp
void calculate(int a, int b, int& sum, int& product)
```

可以透過 `sum` 與 `product` 修改外部的兩個變數。

### C 複合指定

```c
a += 1;
```

等價於：

```c
a = a + 1;
```

類似的寫法還包括：

```text
-=   *=   /=
```
