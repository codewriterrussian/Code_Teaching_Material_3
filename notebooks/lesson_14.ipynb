```python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int a[2][3] = {
        {1, 2, 3},
        {4, 5, 6}
    };

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j < 3; j++) {
            cout << a[i][j] << " ";
        }

        cout << endl;
    }

    return 0;
}
```

```python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.push_back(40);
    a.push_back(50);

    cout << a.size() << endl;

    for (size_t i = 0; i < a.size(); i++) {
        cout << a[i] << " ";
    }

    return 0;
}
```
