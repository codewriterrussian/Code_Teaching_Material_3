# 1. 什麼是陣列？

陣列（Array）可以一次儲存多個 **相同型態** 的資料。

例如要儲存 5 位學生的成績：

```cpp
int score[5] = {80, 90, 75, 88, 95};
```

這裡：

- `int`：元素的型態
- `score`：陣列名稱
- `[5]`：陣列有 5 個元素
- `{...}`：初始值

陣列索引從 **0** 開始：

```text
索引：   0   1   2   3   4
資料：  80  90  75  88  95
```


## 2. 使用索引讀取與修改元素

使用：

```cpp
score[0]
```

可以取得第一個元素。

也可以直接修改：

```cpp
score[0] = 100;
```


````python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int score[5] = {80, 90, 75, 88, 95};

    cout << score[0] << endl;

    score[0] = 100;

    cout << score[0] << endl;

    return 0;
}
````

執行結果：

```text
80
100
```

注意：

```cpp
score[0]
```

是第一個元素，而不是 `score[1]`。


# 3. 使用 `for` 走訪陣列

如果要輸出陣列中的所有元素，不需要一個一個寫：

```cpp
cout << a[0];
cout << a[1];
cout << a[2];
...
```

可以使用 `for` 迴圈：

```cpp
for (int i = 0; i < 5; i++) {
    cout << a[i] << endl;
}
```


````python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int a[5] = {10, 20, 30, 40, 50};

    for (int i = 0; i < 5; i++) {
        cout << a[i] << endl;
    }

    return 0;
}
````

這裡的 `i` 依序是：

```text
0 1 2 3 4
```

因此程式依序存取：

```cpp
a[0]
a[1]
a[2]
a[3]
a[4]
```

條件必須寫：

```cpp
i < 5
```

而不是：

```cpp
i <= 5
```

因為 `a[5]` 已經超出陣列範圍。


# 4. 使用 `cin` 輸入陣列

如果陣列中的資料由使用者輸入，可以先宣告：

```cpp
int a[5];
```

接著用 `for` 搭配 `cin`：

```cpp
for (int i = 0; i < 5; i++) {
    cin >> a[i];
}
```


````python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int a[5];

    for (int i = 0; i < 5; i++) {
        cin >> a[i];
    }

    for (int i = 0; i < 5; i++) {
        cout << a[i] << " ";
    }

    return 0;
}
````

例如輸入：

```text
10 20 30 40 50
```

輸出：

```text
10 20 30 40 50
```

第一個 `for` 負責 **輸入**，第二個 `for` 負責 **輸出**。


## 5. 陣列宣告要注意什麼？

以下是正確的宣告：

```cpp
int a[5];
```

也可以在宣告時直接指定每個元素：

```cpp
int a[5] = {10, 20, 30, 40, 50};
```

但下面這種寫法不是「把整個陣列設成 100」：

```cpp
int a[5] = {100};
```

它代表：

```text
a[0] = 100
a[1] = 0
a[2] = 0
a[3] = 0
a[4] = 0
```

如果要把所有元素改成某個值，通常需要使用迴圈。


# 6. 二維陣列

如果資料像表格一樣有「列」和「欄」，可以使用二維陣列。

例如：

```cpp
int a[2][3] = {
    {1, 2, 3},
    {4, 5, 6}
};
```

可以想成：

| | 第 0 欄 | 第 1 欄 | 第 2 欄 |
|---|---:|---:|---:|
| 第 0 列 | 1 | 2 | 3 |
| 第 1 列 | 4 | 5 | 6 |

存取方式：

```cpp
a[列][欄]
```

例如：

```cpp
a[1][2]
```

得到 `6`。


````python
%%cpp

#include <iostream>
using namespace std;

int main() {
    int a[2][3] = {
        {1, 2, 3},
        {4, 5, 6}
    };

    for (int i = 0; i < 2; i++) {
        for (int j = 0; j < 3; j++) {
            cout << a[i][j] << " ";
        }

        cout << endl;
    }

    return 0;
}
````

二維陣列通常會使用 **巢狀迴圈**：

```cpp
for (int i = 0; i < 列數; i++) {
    for (int j = 0; j < 欄數; j++) {
        ...
    }
}
```

- `i`：控制第幾列
- `j`：控制第幾欄


# 7. 傳統陣列的限制

傳統陣列建立後，大小通常固定。

例如：

```cpp
int a[5];
```

表示這個陣列有 5 個位置。

不能像 Python 的 `list`：

```python
lst.append(10)
```

一樣直接把陣列擴充成更多元素。

另外，C++ 傳統陣列中的元素必須是相同型態，例如：

```cpp
int a[5];
```

裡面的元素都是 `int`。

因此如果需要一個可以方便增加、刪除元素的容器，C++ 常使用 `vector`。


# 8. C++ `vector`

使用 `vector` 前需要：

```cpp
#include <vector>
```

例如：

```cpp
vector<int> a = {10, 20, 30};
```

代表建立一個存放 `int` 的 `vector`。

和陣列一樣，索引也是從 `0` 開始。


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    cout << a[0] << endl;

    return 0;
}
````

# 9. `push_back()`：在尾端加入元素

`vector` 可以使用：

```cpp
a.push_back(40);
```

把 `40` 加到最後面。

原本：

```text
10 20 30
```

加入後：

```text
10 20 30 40
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.push_back(40);
    a.push_back(50);

    for (size_t i = 0; i < a.size(); i++) {
        cout << a[i] << " ";
    }

    return 0;
}
````

# 10. `size()`：取得元素數量

`vector` 可以使用：

```cpp
a.size()
```

取得目前有多少個元素。

例如：

```cpp
vector<int> a = {10, 20, 30};

a.push_back(40);
a.push_back(50);
```

此時：

```cpp
a.size()
```

會得到 `5`。


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.push_back(40);
    a.push_back(50);

    cout << "元素數量：" << a.size() << endl;

    for (size_t i = 0; i < a.size(); i++) {
        cout << a[i] << " ";
    }

    return 0;
}
````

## 為什麼這裡使用 `size_t`？

常見寫法：

```cpp
for (size_t i = 0; i < a.size(); i++)
```

因為 `a.size()` 回傳的是一種 **無號整數型態**，通常使用 `size_t` 來搭配。

如果寫：

```cpp
for (int i = 0; i < a.size(); i++)
```

某些編譯器可能會出現：

```text
comparison of integers of different signs
```

的警告。

初學時可以先記住：

> 用索引搭配 `vector.size()` 時，`size_t` 是很常見的寫法。


# 11. `pop_back()`：刪除最後一個元素

```cpp
a.pop_back();
```

會刪除 `vector` 最後面的元素。

例如：

```text
10 20 30
```

執行：

```cpp
a.pop_back();
```

後變成：

```text
10 20
```


````python
%%cpp

#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> a = {10, 20, 30};

    a.pop_back();

    for (size_t i = 0; i < a.size(); i++) {
        cout << a[i] << " ";
    }

    return 0;
}
````

# 12. Array 與 `vector` 比較

| 特性 | 傳統 Array | `vector` |
|---|---|---|
| 元素型態 | 相同 | 相同 |
| 使用索引 | ✅ | ✅ |
| 索引從 0 開始 | ✅ | ✅ |
| 大小容易改變 | ❌ | ✅ |
| `push_back()` | ❌ | ✅ |
| `pop_back()` | ❌ | ✅ |
| `size()` | 不直接提供相同用法 | ✅ |

如果題目已經知道固定大小，傳統陣列可以很方便。

如果資料數量可能改變，通常 `vector` 會更方便。

# 小結

這份筆記最重要的觀念：

```text
Array
│
├─ 固定大小
├─ 相同型態
├─ 使用索引 a[i]
└─ 常搭配 for 迴圈
      │
      └─ 二維陣列 → 巢狀 for

vector
│
├─ 相同型態
├─ 使用索引 a[i]
├─ push_back() 加入元素
├─ pop_back() 刪除最後元素
└─ size() 取得元素數量
```

初學時可以先掌握：

```cpp
int a[5];
```

與：

```cpp
vector<int> a;
```

兩者在「大小是否方便改變」上的差別。
