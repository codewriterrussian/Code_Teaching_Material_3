````python
# === C++ Vector 自動檢查工具：請先執行這個 cell，不要修改 ===
import base64
import json
import os
import shutil
import subprocess
import tempfile
from IPython.core.magic import register_cell_magic

_TEST_CASES = json.loads(
    base64.b64decode("eyJRMSI6W3siaW5wdXQiOiI1XG4zIDUgMCAtMiA4XG4iLCJleHBlY3RlZCI6IjMgNSAwIC0yIDhcbiJ9LHsiaW5wdXQiOiIxXG40MlxuIiwiZXhwZWN0ZWQiOiI0MlxuIn0seyJpbnB1dCI6IjZcbi0xIC0yIC0zIDAgNCA5XG4iLCJleHBlY3RlZCI6Ii0xIC0yIC0zIDAgNCA5XG4ifV0sIlEyIjpbeyJpbnB1dCI6IjVcbjEwIDIwIDMwIDQwIDUwXG4iLCJleHBlY3RlZCI6IjVcbiJ9LHsiaW5wdXQiOiIxXG43XG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiI4XG4xIDEgMiAzIDUgOCAxMyAyMVxuIiwiZXhwZWN0ZWQiOiI4XG4ifV0sIlEzIjpbeyJpbnB1dCI6IjVcbjEwIDIwIDMwIDQwIDUwXG4iLCJleHBlY3RlZCI6IjEwIDUwXG4ifSx7ImlucHV0IjoiMVxuLTdcbiIsImV4cGVjdGVkIjoiLTcgLTdcbiJ9LHsiaW5wdXQiOiI0XG45IDAgLTIgNlxuIiwiZXhwZWN0ZWQiOiI5IDZcbiJ9XSwiUTQiOlt7ImlucHV0IjoiNVxuMTAgMjAgMzAgNDAgNTBcbjIgOTlcbiIsImV4cGVjdGVkIjoiMTAgMjAgOTkgNDAgNTBcbiJ9LHsiaW5wdXQiOiIzXG4xIDIgM1xuMCAtNVxuIiwiZXhwZWN0ZWQiOiItNSAyIDNcbiJ9LHsiaW5wdXQiOiI0XG44IDggOCA4XG4zIDFcbiIsImV4cGVjdGVkIjoiOCA4IDggMVxuIn1dLCJRNSI6W3siaW5wdXQiOiI1XG4xIDIgMyA0IDVcbiIsImV4cGVjdGVkIjoiNSA0IDMgMiAxXG4ifSx7ImlucHV0IjoiMVxuNDJcbiIsImV4cGVjdGVkIjoiNDJcbiJ9LHsiaW5wdXQiOiI2XG4zIDAgLTEgOSA5IDJcbiIsImV4cGVjdGVkIjoiMiA5IDkgLTEgMCAzXG4ifV0sIlE2IjpbeyJpbnB1dCI6IjNcbjEwIDIwIDMwXG4yXG40MCA1MFxuIiwiZXhwZWN0ZWQiOiIxMCAyMCAzMCA0MCA1MFxuNVxuIn0seyJpbnB1dCI6IjBcblxuM1xuNyA4IDlcbiIsImV4cGVjdGVkIjoiNyA4IDlcbjNcbiJ9LHsiaW5wdXQiOiI0XG4xIDEgMiAzXG4wXG5cbiIsImV4cGVjdGVkIjoiMSAxIDIgM1xuNFxuIn1dLCJRNyI6W3siaW5wdXQiOiI1XG4xMCAyMCAzMCA0MCA1MFxuMlxuIiwiZXhwZWN0ZWQiOiIxMCAyMCAzMFxuM1xuIn0seyJpbnB1dCI6IjNcbjcgOCA5XG4zXG4iLCJleHBlY3RlZCI6IkVNUFRZXG4wXG4ifSx7ImlucHV0IjoiNFxuMSAtMSAyIC0yXG4wXG4iLCJleHBlY3RlZCI6IjEgLTEgMiAtMlxuNFxuIn1dLCJROCI6W3siaW5wdXQiOiI0XG4xMCAyMCAzMCA0MFxuIiwiZXhwZWN0ZWQiOiIwXG5FTVBUWVxuIn0seyJpbnB1dCI6IjFcbi01XG4iLCJleHBlY3RlZCI6IjBcbkVNUFRZXG4ifSx7ImlucHV0IjoiNlxuMSAyIDMgNCA1IDZcbiIsImV4cGVjdGVkIjoiMFxuRU1QVFlcbiJ9XSwiUTkiOlt7ImlucHV0IjoiNVxuMSAyIDMgNCA1XG4iLCJleHBlY3RlZCI6IjE1XG4ifSx7ImlucHV0IjoiNFxuLTUgLTIgNyAwXG4iLCJleHBlY3RlZCI6IjBcbiJ9LHsiaW5wdXQiOiI2XG4xMCAyMCAzMCAtMTAgLTIwIC0zMFxuIiwiZXhwZWN0ZWQiOiIwXG4ifV0sIlExMCI6W3siaW5wdXQiOiI2XG4xIDIgMyA0IDYgOVxuIiwiZXhwZWN0ZWQiOiIzXG4ifSx7ImlucHV0IjoiNVxuMSAzIDUgNyA5XG4iLCJleHBlY3RlZCI6IjBcbiJ9LHsiaW5wdXQiOiI3XG4wIC0yIC0zIDggMTEgMTQgMTVcbiIsImV4cGVjdGVkIjoiNFxuIn1dLCJRMTEiOlt7ImlucHV0IjoiNlxuMyA5IDUgOSAxIDhcbiIsImV4cGVjdGVkIjoiOSAxXG4ifSx7ImlucHV0IjoiNVxuLTggLTMgLTExIC00IC0zXG4iLCJleHBlY3RlZCI6Ii0zIDFcbiJ9LHsiaW5wdXQiOiI0XG43IDYgNSA0XG4iLCJleHBlY3RlZCI6IjcgMFxuIn1dLCJRMTIiOlt7ImlucHV0IjoiNVxuMTAgMjAgMzAgMjAgNDBcbjIwXG4iLCJleHBlY3RlZCI6IjFcbiJ9LHsiaW5wdXQiOiI0XG4xIDIgMyA0XG45XG4iLCJleHBlY3RlZCI6Ii0xXG4ifSx7ImlucHV0IjoiNlxuLTEgMCAtMSAyIC0xIDNcbi0xXG4iLCJleHBlY3RlZCI6IjBcbiJ9XSwiUTEzIjpbeyJpbnB1dCI6IjZcbjEgMiAyIDMgMiA0XG4yIDlcbiIsImV4cGVjdGVkIjoiMSA5IDkgMyA5IDRcbiJ9LHsiaW5wdXQiOiI1XG4wIDAgMCAwIDBcbjAgLTFcbiIsImV4cGVjdGVkIjoiLTEgLTEgLTEgLTEgLTFcbiJ9LHsiaW5wdXQiOiI0XG41IDYgNyA4XG45IDFcbiIsImV4cGVjdGVkIjoiNSA2IDcgOFxuIn1dLCJRMTQiOlt7ImlucHV0IjoiNVxuMSAyIDMgNCA1XG4xMFxuIiwiZXhwZWN0ZWQiOiIxMCAyMCAzMCA0MCA1MFxuIn0seyJpbnB1dCI6IjRcbi0yIDAgMyA3XG4tMVxuIiwiZXhwZWN0ZWQiOiIyIDAgLTMgLTdcbiJ9LHsiaW5wdXQiOiIzXG41IDUgNVxuMFxuIiwiZXhwZWN0ZWQiOiIwIDAgMFxuIn1dLCJRMTUiOlt7ImlucHV0IjoiN1xuLTMgMCA1IDIgLTEgOCAwXG4iLCJleHBlY3RlZCI6IjUgMiA4XG4zXG4ifSx7ImlucHV0IjoiNFxuLTEgMCAtMiAwXG4iLCJleHBlY3RlZCI6IkVNUFRZXG4wXG4ifSx7ImlucHV0IjoiNVxuMSAyIDMgNCA1XG4iLCJleHBlY3RlZCI6IjEgMiAzIDQgNVxuNVxuIn1dLCJRMTYiOlt7ImlucHV0IjoiMiAzXG4xIDIgM1xuNCA1IDZcbiIsImV4cGVjdGVkIjoiMSAyIDNcbjQgNSA2XG4ifSx7ImlucHV0IjoiMSA0XG45IDggNyA2XG4iLCJleHBlY3RlZCI6IjkgOCA3IDZcbiJ9LHsiaW5wdXQiOiIzIDJcbi0xIDBcbjIgM1xuNCAtNVxuIiwiZXhwZWN0ZWQiOiItMSAwXG4yIDNcbjQgLTVcbiJ9XSwiUTE3IjpbeyJpbnB1dCI6IjIgM1xuMSAyIDNcbjQgNSA2XG4iLCJleHBlY3RlZCI6IjYgMTVcbiJ9LHsiaW5wdXQiOiIzIDJcbi0xIDFcbjEwIC0zXG4wIDBcbiIsImV4cGVjdGVkIjoiMCA3IDBcbiJ9LHsiaW5wdXQiOiIxIDVcbjEgMSAxIDEgMVxuIiwiZXhwZWN0ZWQiOiI1XG4ifV0sIlExOCI6W3siaW5wdXQiOiIyIDNcbjEgMiAzXG40IDUgNlxuIiwiZXhwZWN0ZWQiOiI1IDcgOVxuIn0seyJpbnB1dCI6IjMgMlxuLTEgMVxuMTAgLTNcbjAgNVxuIiwiZXhwZWN0ZWQiOiI5IDNcbiJ9LHsiaW5wdXQiOiIxIDRcbjIgNCA2IDhcbiIsImV4cGVjdGVkIjoiMiA0IDYgOFxuIn1dLCJRMTkiOlt7ImlucHV0IjoiMiAzXG4xIDIgM1xuNCA1IDZcbiIsImV4cGVjdGVkIjoiMSA0XG4yIDVcbjMgNlxuIn0seyJpbnB1dCI6IjEgNFxuOSA4IDcgNlxuIiwiZXhwZWN0ZWQiOiI5XG44XG43XG42XG4ifSx7ImlucHV0IjoiMyAyXG4tMSAwXG4yIDNcbjQgLTVcbiIsImV4cGVjdGVkIjoiLTEgMiA0XG4wIDMgLTVcbiJ9XSwiUTIwIjpbeyJpbnB1dCI6IjIgM1xuM1xuMCAwIDVcbjEgMiA5XG4wIDEgN1xuIiwiZXhwZWN0ZWQiOiI1IDcgLTFcbi0xIC0xIDlcbiJ9LHsiaW5wdXQiOiIzIDJcbjJcbjIgMSA0XG4xIDAgMFxuIiwiZXhwZWN0ZWQiOiItMSAtMVxuMCAtMVxuLTEgNFxuIn0seyJpbnB1dCI6IjEgNFxuNFxuMCAwIDFcbjAgMSAyXG4wIDIgM1xuMCAzIDRcbiIsImV4cGVjdGVkIjoiMSAyIDMgNFxuIn1dfQ==").decode("utf-8")
)

_ANSWERS = {}
_RESULTS = {}

def _normalise_output(text):
    lines = [line.rstrip() for line in text.replace("\r\n", "\n").split("\n")]
    while lines and lines[-1] == "":
        lines.pop()
    return "\n".join(lines)

def _find_cpp_compiler():
    for name in ["g++", "clang++", "c++"]:
        path = shutil.which(name)
        if path:
            return path
    return None

def _compile_and_run(source, input_text):
    compiler = _find_cpp_compiler()

    if compiler is None:
        return False, "", "找不到 C++ 編譯器（g++ / clang++ / c++）。"

    with tempfile.TemporaryDirectory() as td:
        src = os.path.join(td, "answer.cpp")
        exe = os.path.join(td, "answer")
        compile_cmd = [compiler, "-std=c++17", src, "-o", exe]

        with open(src, "w", encoding="utf-8") as f:
            f.write(source)

        compile_result = subprocess.run(
            compile_cmd,
            capture_output=True,
            text=True,
        )

        if compile_result.returncode != 0:
            return False, "", compile_result.stderr

        try:
            run_result = subprocess.run(
                [exe],
                input=input_text,
                capture_output=True,
                text=True,
                timeout=5,
            )
        except subprocess.TimeoutExpired:
            return False, "", "程式執行超過 5 秒，請檢查是否有無限迴圈。"

        if run_result.returncode != 0:
            err = run_result.stderr or f"程式結束碼：{run_result.returncode}"
            return False, run_result.stdout, err

        return True, run_result.stdout, ""

def _save_answer(qid, cell):
    qid = qid.strip().upper()
    if qid not in _TEST_CASES:
        print(f"❌ 未知題號：{qid}")
        return

    _ANSWERS[qid] = cell
    print(f"已儲存 {qid} 的 C++ 作答。請執行下方檢查 cell。")

@register_cell_magic
def cpp_answer(line, cell):
    _save_answer(line, cell)

def _run_check(qid):
    qid = qid.upper()

    if qid not in _ANSWERS:
        print(f"❌ {qid}：尚未執行 C++ 作答 cell")
        _RESULTS[qid] = False
        return

    source = _ANSWERS[qid]
    cases = _TEST_CASES[qid]
    passed = 0

    for case_number, case in enumerate(cases, start=1):
        ok_run, stdout, error = _compile_and_run(
            source,
            case["input"],
        )

        if not ok_run:
            print(f"❌ {qid}：C++ 編譯或執行失敗")
            print(error)
            _RESULTS[qid] = False
            return

        actual = _normalise_output(stdout)
        expected = _normalise_output(case["expected"])

        if actual == expected:
            passed += 1
        else:
            print(f"❌ {qid} 測試 {case_number} 未通過")
            print("測試輸入：")
            print(case["input"], end="")
            print("你的輸出：")
            print(stdout if stdout else "<沒有輸出>")
            print("預期輸出：")
            print(case["expected"], end="")
            print()

    success = passed == len(cases)
    _RESULTS[qid] = success

    if success:
        print(f"✅ {qid} 全部通過（{passed} / {len(cases)}，C++）")
    else:
        print(f"❌ {qid}：通過 {passed} / {len(cases)} 個測試（C++）")

def _final_report():
    passed = sum(_RESULTS.get(f"Q{i}", False) for i in range(1, 21))
    untouched = [f"Q{i}" for i in range(1, 21) if f"Q{i}" not in _RESULTS]

    print(f"總成績：{passed} / 20 題通過")
    if untouched:
        print("尚未執行檢查：" + ", ".join(untouched))

print("C++ Vector 自動檢查工具已載入。")
print("每題先執行 %%cpp_answer Qn 作答 cell，再執行下方檢查 cell。")
print("作答後不需要手動輸入測試資料。")
````
