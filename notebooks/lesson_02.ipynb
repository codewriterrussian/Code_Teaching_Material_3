## 1. 五位數反轉

例如：

```text
12345 → 54321
```

原數字可表示為：

```text
12345 = 1 × 10000 + 2 × 1000 + 3 × 100 + 4 × 10 + 5
```


````python
def q2_reverse_five_digits(number):
    """反轉一個五位正整數。"""
    ten_thousands = number // 10000
    thousands = (number // 1000) % 10
    hundreds = (number // 100) % 10
    tens = (number // 10) % 10
    ones = number % 10

    return (
        ones * 10000
        + tens * 1000
        + hundreds * 100
        + thousands * 10
        + ten_thousands
    )
````

````python
q2_reverse_five_digits(12345)
````

## 3. BMI 計算

BMI 公式：

\[
\text{BMI} =
\frac{\text{體重（公斤）}}
{\text{身高（公尺）}^2}
\]

函式接收公分與公斤，並將結果格式化至小數點後兩位。


````python
def q8_bmi_report(height_cm, weight_kg):
    """根據身高（公分）與體重（公斤）回傳 BMI 報告。"""
    height_m = height_cm / 100
    bmi = weight_kg / (height_m ** 2)

    return f"BMI={bmi:.2f}"
````

````python
q8_bmi_report(180, 82)
````

## 5. 函式的參數、輸出與 `None`

函式可以理解為：

```text
輸入 → 執行 → 輸出
```

`print()` 只會顯示內容，不會自動將該內容作為函式的回傳值。  
沒有使用 `return` 指定值時，函式會回傳 `None`。


````python
def print_out():
    """顯示三次 hi；此函式沒有明確的回傳值。"""
    print("hi" * 3)
````

````python
result = print_out()
print(result)
````
