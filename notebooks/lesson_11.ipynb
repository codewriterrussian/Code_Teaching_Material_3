# C/C++ 函式使用：`void` 與 `return`

## 1. 函式可以有「回傳型態」

C/C++ 函式名稱前面的型態，表示這個函式執行完後會回傳什麼型態的資料。

例如：

```cpp
int add(...)
```

表示函式最後應該回傳一個 `int`。

而：

```cpp
void show(...)
```

表示這個函式**不回傳一個值**。

---

### 原筆記的 `void -> (None)`

這可以當成課堂上的簡短記憶方式，但在 C/C++ 中要更精確地說：

```text
void
↓
這個函式沒有回傳值
```

`void` 並不是 Python 的 `None`。

C/C++ 沒有把 `void` 函式的結果當成一個叫做 `None` 的物件。


## 2. `void` 函式

### 補充說明

下面是用來解釋原筆記概念的簡單例子：

```cpp
void sayHello()
{
    cout << "Hello";
}
```

這個函式的工作是：

```text
呼叫 sayHello()
        ↓
印出 Hello
        ↓
函式結束
```

它有做事情，但沒有把一個數值回傳給呼叫它的地方。

因此：

```cpp
void sayHello()
```

的 `void` 可以理解成：

> 這個函式沒有回傳值。


## 3. `return 0;` 的意思

原筆記另一行是：

```text
return 0 -> return None
```

如果是在 C/C++ 中，這句需要特別修正：

```cpp
return 0;
```

並不是「return None」。

它真正的意思是：

```text
把整數 0 回傳給呼叫這個函式的地方
```

因此，如果函式宣告為：

```cpp
int functionName()
```

那麼：

```cpp
return 0;
```

就是回傳一個 `int`，而它的值剛好是 `0`。


## 4. 常見的 `main()` 例子

### 補充說明

```cpp
int main()
{
    return 0;
}
```

可以拆成兩部分看：

```cpp
int main()
```

表示：

```text
main 是一個回傳 int 的函式
```

而：

```cpp
return 0;
```

表示：

```text
main 最後回傳整數 0
```

所以這裡：

```text
0 是一個真正的 int 值
```

而不是：

```text
None
```


## 5. `void` 與 `return 0;` 不要混在一起

| 寫法 | 意思 |
|---|---|
| `void f()` | `f` 沒有回傳值 |
| `return;` | 結束目前函式，不回傳一個值 |
| `int f()` | `f` 應該回傳一個 `int` |
| `return 0;` | 回傳整數 `0` |

因此：

```cpp
void f()
{
    return;
}
```

和：

```cpp
int f()
{
    return 0;
}
```

是兩種不同的情況。

---

### 重要觀念

不要把：

```text
void
```

記成：

```text
回傳 None
```

比較好的記法是：

```text
void = no return value
```

而：

```text
return 0;
```

則是：

```text
return an integer whose value is 0
```


## 6. 函式執行流程

可以先用最基本的流程理解：

```text
caller
  │
  │ 呼叫函式
  ▼
function
  │
  │ 執行函式內容
  ▼
return / 函式結束
  │
  ▼
caller 繼續執行
```

差別在於：

```text
void function
→ 回去，但沒有帶一個值回去
```

而：

```text
non-void function
→ 可以透過 return 帶一個值回去
```
